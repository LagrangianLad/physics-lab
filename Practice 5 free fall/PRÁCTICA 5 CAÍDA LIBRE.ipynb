## 1. INTRODUCCIÓN Y OBJETIVOS

<p style="text-indent: 2em; text-align: justify;"> Cuando se deja caer un cuerpo desde una determinada altura, este se ve sometido a la fuerza de la gravedad que induce sobre él un movimiento uniformemente acelerado en ausencia de otras fuerzas externas. El estudio de este movimiento es de gran relevancia en la mecánica clásica pues modela muchas situaciones físcas comunes en el día a día. En general, este tipo de movimiento depende únicamente de la fuerza de la gravedad así como de la velocidad inicial que se le induce al cuerpo cuando se deja caer. Normalmente, se toma un sistema de referencia de tal forma que la aceleración de la gravedad se supone con signo negativo y, por tanto, si la velocidad inicial va en su misma dirección, compartirá dicho signo. 

<p style="text-indent: 2em; text-align: justify;"> La importancia de fijar un sistema de referencia adecuado no es nimia, puesto que a la hora de hacer mediciones experimentales en laboratorio, conviene elegirlo de tal forma que minimice el error humano que puede haber cuando se toman medidas. En particular, fijar el instante de tiempo inicial adecuado hará que el tiempo de reacción que interviene al medir tiempos se minimice y, por tanto, no afecte a la calidad del experimento. Asimismo, convendrá pues fijar una altura de referencia que facilite la modelización del experimento; normalmente se toma como tal el suelo del laboratorio aunque en otras ocasiones una elección distinta se efectuará por los motivos mencionados con anterioridad.

<p style="text-indent: 2em; text-align: justify;"> Por consiguiente, esta práctica tiene como objetivos el estudio del movimiento en caída libre de un cuerpo en dos supuestos diferentes: velocidad inicial nula y velocidad inicial no nula. En el primer caso, se estudiará la relación lineal que existe entre el tiempo de caída al cuadrado y la altura sobre la que se deja caer el cuerpo para, mediante una regresión lineal, hallar experimentalmente el valor de la aceleración de la gravedad. Para el segundo supuesto, se hará una regresión no lineal, obteniendo mediante ella el valor de la velocidad inicial que adquirió el cuerpo en el experimento así como un segundo valor de la aceleración de la gravedad. Por tanto, la práctica tiene como objetivos secundarios el manejo de la aplicación audacity para las mediciones de los intervalos de tiempo así como la realización de ajustes lineales y no lineales por mínimos cuadrados.

## 2. MARCO TEÓRICO GENERAL

<p style="text-indent: 2em; text-align: justify;"> Como se ha mencionado en la introducción, el objetivo principal es el estudio de un cuerpo en caída libre. La ecuación que utilizaremos para modelar su movimiento es la siguiente:

$$ z(t) = z_0 -v_0\cdot t - \frac{1}{2}g\cdot t^2 \tag{1}$$

<p style="text-indent: 2em; text-align: justify;">siendo $z_0$, $v_0$ la altura y velocidad incial, respectivamente. El eje de alturas se ha orientado con sus valores positivos hacia arriba, de ahí que $g$ y $v_0$ aparezcan con el signo negativo en la ecuación. Si suponemos $v_0 = 0$ y calculamos el tiempo que tarda el cuerpo en caer al suelo $z = 0$, obtenemos:

$$ t^2 = \frac{2}{g} z_0 \tag{2}$$

<p style="text-indent: 2em; text-align: justify;">De esta forma, obtenemos una relación lineal entre la altura inicial y el tiempo de caída, cuya pendiente viene dada por $ a = \frac{2}{g} $ y su ordenada en el origen $b \approx 0$. Expermientalmente, introduciremos una bola metálica pequeña dentro de un globo, el cual fijaremos a una altura dada con la ayuda de un metro. Cuidadosamente para no inducir velocidad inicial alguna, se pinchará el globo y, con la aplicación Audacity se registrarán los intervalos de tiempo en los que el globo explota y la bola impacta con la plataforma metálica que se encuentra bajo el metro. Así determinaremos $t$, y $z_0$ será la altura a la que se pone el globo.

<p style="text-indent: 2em; text-align: justify;">Para la experiencia de caída libre con velocidad inicial no nula, consideremos el movimiento de una bola-imán que cae a través de un tubo rodeado de expiras de cobre. Si $z_0'$ es la posición de vertical de la primera espira con respecto a la base del tubo, $v_0$ la velocidad de la bola al pasar por dicha espira y $t_0'$ el instante de tiempo en el que atraviesa la primera espira, entonces:

$$ z'(t') = z_0' +v_0\cdot (t'-t_0) + \frac{1}{2}g \cdot {(t'-t_0')}^2 \tag{3}$$

<p style="text-indent: 2em; text-align: justify;">donde se ha tomado el eje de alturas orientado con sus valores positivos hacia abajo. Como origen de alturas, se tomará la primera espira $z_0' = 0$, de manera que la altura resulta $z = z' - z_0'$. En cuanto al origen de tiempos, se elige el insante $t_0'$ como origen, por tanto, $t = t' -t_0'$. Si escribimos la inversa de $(3)$:

$$ t = \frac{-v_0 + \sqrt{v_0^2 + 2\cdot g \cdot z}}{g} \tag{4}$$

<p style="text-indent: 2em; text-align: justify;">obtenemos la ecuación buscada para realizar el ajuste no lineal de $t$ frente a $z$ para obtener $v_0$ y $g$

$$ u_{t^2} = 2\cdot \overline{t^2}\cdot \frac{u_t}{\overline t} = 2\cdot \overline t \cdot u_t$$

```python
t1 = np.array([0.604, 0.557, 0.497, 0.601, 0.476, 0.437, 0.527, 0.401, 0.304, 0.180]) #s
t2 = np.array([0.597, 0.552, 0.488, 0.592, 0.469, 0.435, 0.528, 0.401, 0.305, 0.183]) #s
t3 = np.array([0.605, 0.580, 0.490, 0.593, 0.472, 0.435, 0.518, 0.399, 0.305, 0.182]) #s
t4 = np.array([0.635, 0.576, 0.489, 0.594, 0.470, 0.436, 0.528, 0.403, 0.306, 0.181]) #s
t5 = np.array([0.603, 0.557, 0.485, 0.590, 0.471, 0.434, 0.530, 0.401, 0.304, 0.180]) #s
z_0 = np.array([176.2, 145.5, 117.2, 166.5, 105.0, 91.5, 134.3, 68.2, 43.8, 14.9]) #cm
z_0_metros = z_0/100

def u_b(resolucion, digital : bool = True):
    return resolucion/np.sqrt(12) if digital else resolucion/np.sqrt(24) 
def u_a(s, N):
    return s/ N**0.5
def u_combinada (u_a, u_b):
    return np.sqrt(pow(u_a, 2)+ pow(u_b, 2))
def imprimir (lista, unidades, titulo = ""):
    print(titulo)
    for i in range(len(lista)):
        print(f"{i+1} : {lista[i]} {unidades}")

tiempos_medios = np.array([(t1[i] + t2[i]+ t3[i] +t4[i]+t5[i])/5 for i in range(len(t1))])
tiempos_medios_cuadrado = np.array([t**2 for t in tiempos_medios])
u_a_tiempos = []
for i in range(len(t1)):
    s = np.std([t1[i], t2[i], t3[i], t4[i], t5[i]], ddof = 1)
    u_a_tiempo = u_a(s, 5)
    u_a_tiempos.append(u_a_tiempo)
u_tiempos = np.array([np.sqrt(pow(u_a_tiempos[i], 2) + pow(u_b(0.001), 2)) for i in range(len(t1))])
u_tiempos_cuadrado = np.array([2*tiempos_medios[i]*u_tiempos[i] for i in range(len(t1))])
imprimir(tiempos_medios_cuadrado, "s^2", titulo = "Tiempos medios al cuadrado")
imprimir(u_tiempos_cuadrado, "s^2", titulo = "Incertidumbres tiempos medios cuadrado")
```

```python
#Regresión lineal
def Lineal (parameters, x):
    A, B = parameters
    y = np.multiply(A, x) + B
    return y 
lin_model = Model(Lineal)
data = RealData(z_0_metros, tiempos_medios_cuadrado, sy = u_tiempos_cuadrado)
odr = ODR(data, lin_model, beta0 = [0., 1.])
out = odr.run()
out.pprint()
```

```python
fit_y = Lineal (out.beta, z_0_metros)
plt.errorbar(z_0_metros, tiempos_medios_cuadrado, yerr = u_tiempos_cuadrado, linestyle = "None", marker = "o", label = r"$u_{t^2}$")
plt.plot(z_0_metros, fit_y, label = rf'Ajuste ODR: $t^2$ = {out.beta[0]:.3f}·$z_0$ + {out.beta[1]:.3f}')
plt.plot ([], [], ' ', label = r"a = 0.2098 $\pm$ 0.0050 $s^2$/m")
plt.plot ([], [], ' ',label = r"b = 0.0027 $\pm$ 0.0059 $s^2$")
plt.title (r"Experiencia 1: Ajuste $t^2$ frente a z_0")
plt.xlabel ("z_0 (m)")
plt.ylabel (r"$t^2$ ($s^2$)")
plt.grid()
plt.legend()
plt.show()
```

```python
i = np.arange(1, 11)

z_prime = np.array([10.0, 18.9, 28.1, 36.8, 45.9, 54.9, 64.2, 73.1, 82.1, 91.2]) / 100
z = z_prime - z_prime[0] # Al ser arrays, la resta se aplica a todos los elementos

t1_prime = np.array([0.957, 1.012, 1.055, 1.090, 1.122, 1.151, 1.179, 1.204, 1.227, 1.249])
t2_prime = np.array([2.401, 2.458, 2.502, 2.538, 2.570, 2.599, 2.626, 2.650, 2.674, 2.696])
t3_prime = np.array([2.708, 2.765, 2.809, 2.845, 2.877, 2.906, 2.933, 2.958, 2.981, 3.004])

t1 = t1_prime - t1_prime[0]
t2 = t2_prime - t2_prime[0]
t3 = t3_prime - t3_prime[0]

# Matriz (3x10) de numpy porque así puedo calcular más rápido usando axis
tiempos = np.array([t1, t2, t3])
t_medio = np.mean(tiempos, axis=0)
```

```python
# n = 3 medidas por espira
u_A_t = np.std(tiempos, axis=0, ddof=1) / np.sqrt(3)

# Incertidumbre Tipo B de t (Resolución/error de lectura en Audacity)
resolucion = 0.001
# np.full_like copia el array a uno igual, en este caso copio t_medio
u_B_t = np.array([resolucion / np.sqrt(12) for i in range(len(t_medio))])
u_C_t = np.sqrt(u_A_t**2 + u_B_t**2)

print("TIEMPOS MEDIOS")
for i in range(len(t_medio)):
    print(f"\t{t_medio[i]:.6f}  +/- {u_C_t[i]:.6f} s")
```

```python
def modelo_caida(z, v0, g): 
    # t = (-v0 + sqrt(v0^2 + 2*g*z)) / g
    v0_cuadrado = np.power(v0, 2)
    dos_g_z = np.multiply(2, np.multiply(g, z))
    raiz = np.sqrt(v0_cuadrado + dos_g_z)
    numerador = -v0 + raiz
    t = np.divide(numerador, g)
    return t

p0 = [1.0, 9.8]
popt, pcov = curve_fit(modelo_caida, z, t_medio, p0=p0, sigma=u_C_t, absolute_sigma=True)

v0_opt = popt[0]
g_opt = popt[1]

error_params = np.sqrt(np.diag(pcov))
v0_err = error_params[0]
g_err = error_params[1]
```

```python

print("RESULTADOS DEL AJUSTE:")
print("Velocidad inicial v0: {:.4f} +/- {:.4f} m/s".format(v0_opt, v0_err))
print("Gravedad g:           {:.4f} +/- {:.4f} m/s^2".format(g_opt, g_err))
```

```python
# Muestro el gráfico
z_linea = np.linspace(min(z), max(z), 100)
t_linea = modelo_caida(z_linea, v0_opt, g_opt)

# 4. Creación del Plot
plt.figure(figsize=(10, 6))

# Puntos con barras de error
plt.errorbar(z, t_medio, yerr=u_C_t, fmt='o', color='black', 
             ecolor='black', capsize=3, label='Medidas')

# Curva de ajuste (CORREGIDO: r antes de las comillas para LaTeX)
plt.plot(z_linea, t_linea, color='blue', 
         label=f'Velocidad inicial v0: {v0_opt:.3f} +/- {round(v0_err,3)} m/s\n Gravedad g:           {g_opt:.2f} +/- {round(g_err, 2)} m/s^2')

plt.title('Experiencia 2: Ajuste No Lineal del Tiempo de Caída frente a Posición', fontsize=14)
plt.xlabel('Posición vertical $z$ (m)', fontsize=12)
plt.ylabel('Tiempo medio $\\bar{t}$ (s)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

plt.show()
```
