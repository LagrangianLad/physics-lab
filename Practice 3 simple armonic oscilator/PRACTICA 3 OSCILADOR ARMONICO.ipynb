## 2. Marco Teórico General

### 2.1. Experiencia estática: Ley de Hooke

<p style="text-indent: 2em; text-align: justify;">De acuerdo con la ley de Hooke, un muelle al que se le aplica una fuerza $F$ experimenta un estiramiento o elongación de forma proporcional a dicha fuerza. Esta relación se expresa matemáticamente mediante la ecuación:
    
$$\Delta l=\frac{1}{k}F$$

<p style="text-indent: 2em; text-align: justify;">Donde $\Delta l=l-l_{0}$ representa la elongación del muelle, siendo $l_{0}$ su longitud de equilibrio (que corresponde a la longitud del muelle con el portapesas colgado) y $l$ la longitud final tras aplicar la fuerza de las pesas adicionales. La constante elástica $k$ es una característica intrínseca del muelle que depende del material y de su geometría. En esta experiencia, la fuerza aplicada corresponde al peso de una masa $M$, cumpliéndose en el equilibrio que $Mg=k\Delta l$.

### 2.2. Experiencia dinámica: Oscilador armónico

<p style="text-indent: 2em; text-align: justify;">Si el muelle, cargado con una masa $M$ en su estado de equilibrio, se estira ligeramente y se suelta, la masa oscila en torno a su posición de equilibrio realizando un movimiento armónico simple. La ecuación diferencial que rige este movimiento es:
    
$$\frac{d^{2}(\Delta l)}{dt^{2}}+\omega^{2}\Delta l=0$$

<p style="text-indent: 2em; text-align: justify;">Donde $\omega^{2}=k/M$. A partir de la relación con el periodo de oscilación $T=2\pi/\omega$, se obtiene que el periodo de un oscilador ideal viene dado por:
    
$$T=2\pi\sqrt{\frac{M}{k}}$$

### 2.3. Influencia de la masa del muelle y determinación de variables
<p style="text-indent: 2em; text-align: justify;">Si planteamos un segundo análisis más preciso, no debemos despreciar la masa del muelle $m^*$, ya que esta contribuye a la masa efectiva que oscila. Para corregir este efecto, se utiliza la relación $M=Nm+\alpha m^*$, donde $N$ es el número de pesas y $\alpha \simeq 1/3$ es un parámetro que ajusta la influencia de la masa del muelle. La ecuación final para el análisis de datos se convierte en:
    
$$T^{2}=\frac{4\pi^{2}m}{k}N+\frac{4\pi^{2}\alpha m^{*}}{k}$$

<p style="text-indent: 2em; text-align: justify;">A través de ajustes lineales por mínimos cuadrados de $\Delta l$ frente a $N$ y de $T^2$ frente a $N$, es posible determinar experimentalmente la constante elástica $k$, la aceleración de la gravedad $g$ y el factor de influencia $\alpha$, comparándolos posteriormente con sus valores nominales.

### 4.3 Cálculo de las constantes elásticas y del parámetro $\alpha$

<p style="text-indent: 2em; text-align: justify;">En esta sección se presentan los valores obtenidos para la constante elástica del muelle ($k$) y el parámetro de masa efectiva ($\alpha$), calculados a partir de los resultados de la regresión lineal de ambas experiencias. En ambas experiencias se ha utilizado el valor de las masas obtenido en el apartado 4.2.1.

  
#### 4.3.1 Constante elástica (experiencia dinámica)

<p style="text-indent: 2em; text-align: justify;">Según la Ecuación (7) del guion, la relación entre el periodo y el número de pesas es $T^2 = \frac{4\pi^2 m}{k}N$. Por tanto, la pendiente $a$ de la recta de regresión permite obtener $k$. Asimismo, la Ecuación (8) permite obtener $\alpha$ a partir de la ordenada en el origen $b$.

<p style="text-indent: 2em; text-align: justify;">Despejamos de la expresión de la pendiente $a = \frac{4\pi^2 m}{k}$ de donde calculamos: 
    
  $$k = \frac{4\pi^2 m}{a} = 5.625 \pm 0.097 N/m$$

<p style="text-indent: 2em; text-align: justify;">Donde la incertidumbre se ha calculado con la fórmula de propagación de incertidumbres para producto de magnitudes. En particular:

$$u_k = \overline{k}\cdot\sqrt{\epsilon_m^2 + \epsilon_a^2} = \overline{k}\cdot\sqrt{\left(\frac{u_m}{\overline{m}}\right)^2 + \left(\frac{u_a}{\overline{a}}\right)^2}$$

<p style="text-indent: 2em; text-align: justify;">Donde $u_a$ es la incertidumbre obtenida por regresión lineal y $u_m$ es la incertidumbre calculada en el apartado 4.2.1. Los cálculos realizados para determinar este valor se han efectuado en el anexo 7.3.


#### 4.3.2 Parámetro de masa efectiva del muelle $\alpha$.
<p style="text-indent: 2em; text-align: justify;">Despejamos de la expresión de la ordenada en el origen $b = \frac{4\pi^2 \alpha m^*}{k}$:
    
  $$\alpha = \frac{b \cdot k}{4\pi^2 m^*} = 0.344 ± 0.067$$

<p style="text-indent: 2em; text-align: justify;">La incertidumbre ha sido calculada de forma análoga al resultado inmediatamente anterior. $$u_{\alpha}   = \overline{\alpha}\cdot\sqrt{\left(\frac{u_k}{\overline{k}}\right)^2 + \left(\frac{u_b}{\overline{b}}\right)^2 + \left(\frac{u_{m^*}}{\overline{m^*}}\right)^2}$$
<p style="text-indent: 2em; text-align: justify;">Donde asumimos que $\pi$ es una constante que podemos determinar con incertidumbre cero, es decir con todos los decimales de precisión que consideremos necesarios. La incertidumbre asociada a $m^*$, su incertidumbre (únicamente tipo B) será la asociada a la balanza digital, que sigue una distribución rectangular, siendo por tanto $u_{m^*} = \frac{0.01}{\sqrt{12}}$. Los cálculos realizados para determinar este valor se han efectuado en el anexo 7.4.

    
#### 4.3.3 Aceleración debida a la gravedad $g$
<p style="text-indent: 2em; text-align: justify;">Una vez calculada la constante elástica del muelle podemos calcular la aceleración debida a la fuerza de la gravedad despejando de la ecuación: 
    
$$\Delta l = \left( \frac{g \cdot m}{k} \right) N$$

<p style="text-indent: 2em; text-align: justify;">Esta expresión representa una relación lineal de $\Delta l$ frente a $N$, donde la pendiente experimental $a$ está definida por:
    
$$a = \frac{g \cdot m}{k}$$

<p style="text-indent: 2em; text-align: justify;">Concluimos entonces:

$$g= 9.95 ± 0.20 m/s^2$$

<p style="text-indent: 2em; text-align: justify;">Donde el valor de la incertidumbre asociada ha sido calculado como:

$$u_{g}   = \overline{g}\cdot\sqrt{\left(\frac{u_k}{\overline{k}}\right)^2 + \left(\frac{u_a}{\overline{a}}\right)^2 + \left(\frac{u_{m}}{\overline{m}}\right)^2}$$

<p style="text-indent: 2em; text-align: justify;">Los valores de las incertidumbres han sido detallados en las dos secciones inmediatamente anteriores.

## 7. Anexos: Cálculos de Python

Para realizar los cálculos cargamos las siguientes librerías

```python
import numpy as np
import math
import matplotlib.pyplot as plt
from __future__ import print_function
from scipy import stats
from scipy.optimize import curve_fit
from scipy.stats import t
from scipy.odr import *
```

### 7.1. Cálculo de la masa y su incertidumbre
En ambas experiencias se han utilizado las mismas medidas para la masa:

```python
masas_g = [20.38, 20.32, 20.15, 19.91, 20.12]
m = (sum(masas_g) / len(masas_g))
print(f"Masa promedio de las pesas: {m} g")
```

```python
error_balanza = 0.01 
u_a = np.std(masas_g, ddof=1) / np.sqrt(len(masas_g))
u_b = error_balanza / np.sqrt(12)

u_m = np.sqrt(u_a**2 + u_b**2)

print(f"Masa promedio m = ({m:.3f} ± {u_m:.3f}) g")
```

### 7.2. Cálculo de la constante elástica: experiencia dinámica

```python
# DATOS DE LA EXPERIENCIA DINÁMICA
N_dinamica = np.array([1, 2, 3, 4, 5, 6])
t20_1 = [8.76, 11.53, 13.54, 15.67, 17.98, 18.88]
t20_2 = [8.60, 11.74, 13.61, 15.75, 17.30, 18.77]
t20_3 = [8.44, 11.72, 13.93, 15.53, 17.40, 18.92]
```

```python
t20_promedio = [(t1 + t2 + t3) / 3 for t1, t2, t3 in zip(t20_1, t20_2, t20_3)]
```

```python
periodo_T = [t / 20 for t in t20_promedio]
periodo_T2 = [t**2 for t in periodo_T]
```

Si hacemos la regresión lineal teniendo en cuenta las tres medidas que hemos tomado:

```python
def lineal_model(parameters, x):
    A, B = parameters
    return A * x + B

t20_varianza = np.var([t20_1, t20_2, t20_3], axis=0, ddof=1)
t20_std = np.sqrt(t20_varianza)
T2varianza = (np.array(t20_promedio) / 200)**2 * t20_varianza # Propagación de las varianzas muestrales
T2std = np.sqrt(T2varianza)

res_dinamica = stats.linregress(N_dinamica, periodo_T2)
beta_auto = [res_dinamica.slope, res_dinamica.intercept]

modelo = Model(lineal_model)
data = RealData(N_dinamica, periodo_T2, sy=T2std)
odr = ODR(data, modelo, beta0=beta_auto)
out = odr.run()
beta_final = out.beta
```

```python
plt.figure(figsize=(10, 7))
plt.errorbar(N_dinamica, periodo_T2, yerr=T2std, fmt='ko', capsize=4, label='Medias')
plt.plot(N_dinamica, lineal_model(beta_final, N_dinamica), 'b--', label=f' $T^2 = {beta_final[0]:.6f}N + {beta_final[1]:.6f}$')

plt.xlabel('Número de pesas (N)')
plt.ylabel('$T^2 (s^2)$')
plt.title('Experiencia dinámica')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

```python
k_experimental = (4 * (math.pi**2) * (m/1000)) / res_dinamica.slope 

error_relativo_m = u_m / m
error_relativo_a = res_dinamica.stderr / res_dinamica.slope
u_k = k_experimental * np.sqrt(error_relativo_m**2 + error_relativo_a**2)

print(f"Constante del muelle, experiencia dinámica: {k_experimental:.3f} ± {u_k:.3f} N/m")
```

### 7.3. Cálculo del parámetro $\alpha$

```python
# Cálculo de alfa
b = res_dinamica.intercept
u_b = res_dinamica.intercept_stderr
alpha = (b * k_experimental) / (4 * (np.pi**2) * (m/1000))
u_alpha = alpha * np.sqrt((u_b/b)**2 + (u_k/k_experimental)**2 + (u_k/k_experimental)**2)

print(f"Valor experimental de alpha: {round(alpha, 3)} ± {u_alpha:.3f}")
```

### 7.4. Calculo de la aceleración gravitatoria: experiencia estática

```python
# DATOS DE LA EXPERIENCIA ESTÁTICA
N = np.array([0, 1, 2, 3, 4, 5])
l_n = [8.6, 12.3, 16.1, 19.4, 23.0, 26.5] 
delta_l = [0.0, 3.7, 7.5, 10.8, 14.4, 17.9]
```

```python
res = stats.linregress(N, delta_l)
print(f"R cuadrado: {res.rvalue**2:.6f}")
print(f"Ordenada en el origen: {res.intercept:.6f} cm")
print(f"Error de la ordenada en el origen: {res.intercept_stderr:.6f} cm")
print(f"Pendiente: {res.slope:.6f} cm")
print(f"Error de la pendiente: {res.stderr:.6f} cm")
```

```python
plt.figure(figsize=(8, 8))
plt.scatter(N, delta_l, color='black', zorder=5, label="Medidas experimentales")
plt.plot(N, res.slope*N + res.intercept, color='blue', linestyle='--', label=f"Δl = {res.slope:.6f} * N + {res.intercept:.6f}")

plt.title('Experiencia Estática', fontsize=14, pad=15)
plt.xlabel('Número de pesas (N)', fontsize=12)
plt.ylabel('Elongación Δl (cm)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.show()
```

```python
g_promedio = ((res.slope / 100) * k_experimental) / (m / 1000)

rel_error_a = res.stderr / res.slope
rel_error_k = u_k / k_experimental
rel_error_m = u_m / m

error_g = g_promedio * np.sqrt(rel_error_a**2 + rel_error_k**2 + rel_error_m**2)

print(f"Aceleración de la gravedad (Estática): {round(g_promedio, 2)} ± {error_g:.2f} m/s^2")
```
