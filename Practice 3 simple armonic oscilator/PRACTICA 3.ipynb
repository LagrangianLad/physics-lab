## 2. Marco Teórico General

### 2.1. Experiencia estática: Ley de Hooke

<p style="text-indent: 2em; text-align: justify;">De acuerdo con la ley de Hooke, un muelle al que se le aplica una fuerza $F$ experimenta un estiramiento o elongación de forma proporcional a dicha fuerza. Esta relación se expresa matemáticamente mediante la ecuación:
    
$$\Delta l=\frac{1}{k}F$$

<p style="text-indent: 2em; text-align: justify;">Donde $\Delta l=l-l_{0}$ representa la elongación del muelle, siendo $l_{0}$ su longitud de equilibrio (que corresponde a la longitud del muelle con el portapesas colgado) e $l$ la longitud final tras aplicar la fuerza de las pesas adicionales. La constante elástica $k$ es una característica intrínseca del muelle que depende del material y de su geometría. En esta experiencia, la fuerza aplicada corresponde al peso de una masa $M$, cumpliéndose en el equilibrio que $Mg=k\Delta l$.

### 2.2. Experiencia dinámica: Oscilador armónico

<p style="text-indent: 2em; text-align: justify;">Si el muelle, cargado con una masa $M$ en su estado de equilibrio, se estira ligeramente y se suelta, la masa oscila en torno a su posición de equilibrio realizando un movimiento armónico simple. La ecuación diferencial que rige este movimiento es:
    
$$\frac{d^{2}(\Delta l)}{dt^{2}}+\omega^{2}\Delta l=0$$

<p style="text-indent: 2em; text-align: justify;">Donde $\omega^{2}=k/M$. A partir de la relación con el periodo de oscilación $T=2\pi/\omega$, se obtiene que el periodo de un oscilador ideal viene dado por:
    
$$T=2\pi\sqrt{\frac{M}{k}}$$

### 2.3. Influencia de la masa del muelle y determinación de variables
<p style="text-indent: 2em; text-align: justify;">Si planteamos un segundo análisis más preciso, no debemos despreciar la masa del muelle $m^*$, ya que esta contribuye a la masa efectiva que oscila. Para corregir este efecto, se utiliza la relación $M=Nm+\alpha m^*$, donde $N$ es el número de pesas y $\alpha \simeq 1/3$ es un parámetro que ajusta la influencia de la masa del muelle. La ecuación final para el análisis de datos se convierte en:
    
$$T^{2}=\frac{4\pi^{2}m}{k}N+\frac{4\pi^{2}\alpha m^{*}}{k}$$

<p style="text-indent: 2em; text-align: justify;">A través de ajustes lineales por mínimos cuadrados de $\Delta l$ frente a $N$ y de $T^2$ frente a $N$, es posible determinar experimentalmente la constante elástica $k$, la aceleración de la gravedad $g$ y el factor de influencia $\alpha$, comparándolos posteriormente con sus valores nominales.

## 7. Anexos: Cálculos de Python

Para realizar los cálculos cargamos las siguientes librerías

```python
import numpy as np
import math
import matplotlib.pyplot as plt
from __future__ import print_function
from scipy import stats
from scipy.optimize import curve_fit
from scipy.stats import t
```

### 7.1. Cálculo de la masa y su incertidumbre
En ambas experiencias se han utilizado las mismas medidas para la masa:

```python
masas_g = [20.38, 20.32, 20.15, 19.91, 20.12]
m = (sum(masas_g) / len(masas_g))
print(f"Masa promedio de las pesas: {m} g")
```

```python
error_balanza = 0.01 
u_a = np.std(masas_g, ddof=1) / np.sqrt(len(masas_g))
u_b = error_balanza / np.sqrt(3)

u_m = np.sqrt(u_a**2 + u_b**2)

print(f"Masa promedio m = ({m:.3f} ± {u_m:.3f}) g")
```

### 7.2. Calculo de la constante elástica: experiencia estática

```python
# DATOS DE LA EXPERIENCIA ESTÁTICA
N = np.array([0, 1, 2, 3, 4, 5])
l_n = [8.6, 12.3, 16.1, 19.4, 23.0, 26.5] 
delta_l = [0.0, 3.7, 7.5, 10.8, 14.4, 17.9]
```

```python
res = stats.linregress(N, delta_l)
print(f"R cuadrado: {res.rvalue**2:.6f}")
print(f"Ordenada en el origen: {res.intercept:.6f} cm")
print(f"Error de la ordenada en el origen: {res.intercept_stderr:.6f} cm")
print(f"Pendiente: {res.slope:.6f} cm")
print(f"Error de la pendiente: {res.stderr:.6f} cm")
```

```python
plt.figure(figsize=(8, 8))
plt.scatter(N, delta_l, color='black', zorder=5, label="Medidas experimentales")
plt.plot(N, res.slope*N + res.intercept, color='royalblue', linestyle='--', label=f"Δl = {res.slope:.6f} * N + {res.intercept:.6f}")

plt.title('Experiencia Estática', fontsize=14, pad=15)
plt.xlabel('Número de pesas (N)', fontsize=12)
plt.ylabel('Elongación Δl (cm)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.show()
```

```python
g = 9.80665  # Gravedad estándar (m/s^2)

# a = (g * m) / k  => k = (g * m) / a siendo a la pendiente
k = (g * (m/1000)) / (res.slope / 100)

error_k = (res.stderr / res.slope) * k

print(f"Constante del muelle, experiencia estática: {k:.3f} ± {error_k:.3f} N/m")
```

#### 7.3. Cálculo de la constante elástica: experiencia dinámica

```python
# DATOS DE LA EXPERIENCIA DINÁMICA
N_dinamica = np.array([1, 2, 3, 4, 5, 6])
t20_1 = [8.76, 11.53, 13.54, 15.67, 17.98, 18.88]
t20_2 = [8.60, 11.74, 13.61, 15.75, 17.30, 18.77]
t20_3 = [8.44, 11.72, 13.93, 15.53, 17.40, 18.92]
```

```python
t20_promedio = [(t1 + t2 + t3) / 3 for t1, t2, t3 in zip(t20_1, t20_2, t20_3)]
```

```python
periodo_T = [t / 20 for t in t20_promedio]
periodo_T2 = [t**2 for t in periodo_T]
```

```python
res_dinamica = stats.linregress(N_dinamica, periodo_T2)
print(f"R cuadrado: {res_dinamica.rvalue**2:.6f}")
print(f"Ordenada en el origen: {res_dinamica.intercept:.6f} s^2")
print(f"Error de la ordenada en el origen: {res_dinamica.intercept_stderr:.6f} s^2")
print(f"Pendiente: {res_dinamica.slope:.6f} s^2")
print(f"Error de la pendiente: {res_dinamica.stderr:.6f} s^2")
```

```python
# 4. GRÁFICA PRELIMINAR (T^2 frente a N)
plt.figure(figsize=(8, 8))
plt.scatter(N_dinamica, periodo_T2, color='black', label="Medidas experimentales")
plt.plot(N_dinamica, res_dinamica.slope*N_dinamica + res_dinamica.intercept, color='royalblue', linestyle='--', label=f"T^² = {res.slope:.6f} * N + {res.intercept:.6f}")

plt.xlabel('Número de pesas (N)', fontsize=12)
plt.ylabel('T² (s²)', fontsize=12)
plt.title('Experiencia dinámica', fontsize=14, pad=15)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

```python
k_experimental = (4 * (math.pi**2) * (m/1000)) / res_dinamica.slope 
error_k = k_experimental * (res_dinamica.stderr / res_dinamica.slope )

print(f"Constante del muelle, experiencia dinámica: {k_experimental:.3f} ± {error_k:.3f} N/m")
```

#### 7.4. Cálculo del parámetro $\alpha$

```python
# Cálculo de alfa
b = res_dinamica.intercept
u_b = res_dinamica.intercept_stderr
alpha = (b * k_experimental) / (4 * (np.pi**2) * (m/1000))
u_alpha = alpha * np.sqrt((u_b/b)**2 + (error_k/k_experimental)**2)

print(f"Valor experimental de alpha: {round(alpha, 4)} ± {u_alpha:.3f}")
```
