## 1. INTRODUCCIÓN Y OBJETIVOS


<p style="text-indent: 2em; text-align: justify;">El péndulo simple ideal consiste en un hilo inextensible de longitud $l$ y masa despreciable, en cuyo extremo cuelga una masa $m$ de tamaño también despreciable. El otro extremo se sujeta a un punto fijo de modo que, al desplazarlo un ángulo $\theta_{0}$ de su posición vertical y soltarlo, comienza a oscilar. En un laboratorio, con las herramientas adecuadas, podemos simular este movimiento y estudiar sus características.

<p style="text-indent: 2em; text-align: justify;">En particular, es intersante reparar en el efecto que tiene el ángulo de apertura $ \theta_{0}$ en las oscilaciones y la rapidez con la que el péndulo se mueve alrededor de su punto de equilibrio. En la teoría, la mayoría de veces se trabaja con ecuaciones que corresponden a aproximaciones de primer orden, por tanto, modelizan los problemas de forma aproximada, distando del comportamiento real del cuerpo. 

<p style="text-indent: 2em; text-align: justify;">Como consecuencia de lo anterior, será notoria la diferencia que existirá cuando se trabaje con ángulos de apertura pequeños y otros más grandes, puesto que deberemos ajustar las fórmulas correspondientes para obtener resultados fidedignos a la experiencia física. A colación de este resultado, cabe resaltar que en las expresiones matemáticas utilizadas no se tiene en cuenta el rozamiento debido al aire y la consiguiente pérdida de energía mecánica que trae consigo. Idealmente, el péndulo mantiene una amplitud de oscilación constante, que no se ve afectada por el paso del tiempo. Sin embargo, en la práctica, sí se ve amortiguado y, por tanto, la amplitud se va reduciendo a cada oscilación. Este fenómeno se hace denotar a mayor es el ángulo $\theta_{0}$.

## 2. Marco teórico general

<p style="text-indent: 2em; text-align: justify;">Como se ha mencionado en el apartado anterior, el movimiento oscilatorio efectuado por el péndulo se rige por el Principio de Conservación de la energía:

$$mgl(1-\cos \theta) + \frac{1}{2}m\left(l\frac{d\theta}{dt}\right)^2 = E_0 \tag{1}$$

<p style="text-indent: 2em; text-align: justify;">Donde el primer término hace referencia a la energía potencial, el segundo término a la energía cinética del péndulo y $E_0$ es la energía total, que viene dada por la expresión $E_0 = mgl(1 - \cos \theta_0)$. Resolviendo la ecuación diferencial obtenemos la expresión aproximada para el periodo de oscilación en función de la amplitud inicial:

$$T \approx 2\pi\sqrt{\frac{l}{g}} \left( 1 + \frac{1}{4}\sin^2\frac{\theta_0}{2} + \frac{9}{64}\sin^4\frac{\theta_0}{2} + \dots \right)~\tag{2}$$ 

<p style="text-indent: 2em; text-align: justify;">Para pequeñas amplitudes, las correcciones debidas a $\theta_{0}$ se pueden despreciar, resultando en un periodo $T$ que depende únicamente de la longitud y la gravedad:

$$T = T_{0} = 2\pi\sqrt{\frac{l}{g}} \tag{3}$$

<p style="text-indent: 2em; text-align: justify;">Lo cual nos permite relacionar de manera lineal las magnitudes $T^2$ y $l$ y, por tanto, hacer un ajuste por mínimos cuadrados, en el que la pendiente nos proporcionará de manera experimental el valor de la gravedad $g$ con su correspondiente incertidumbre $u_g$:

$$ T^2 = \frac{4 \pi^2}{g} l \tag{4}$$

<p style="text-indent: 2em; text-align: justify;">Para la segunda parte del experimento, nos quedaremos con el segundo término en la aproximación de T:

$$ T = T_0\cdot(1 + \frac{1}{4} \sin^2 \frac{\theta_0}{2})$$
$$ \frac{T}{T_0} = 1 + \frac{1}{4} \sin^2 \frac{\theta_0}{2} \tag{5}$$

<p style="text-indent: 2em; text-align: justify;">Ecuación que nos permitirá hacer un ajuste por mínimos cuadrados de $\frac{T}{T_0}$ frente a $\sin^2 \frac{\theta_0}{2}$. La pendiente tendrá que ser aproximadamente $\frac{1}{4}$ y la ordenada en el origen, $1$. De esta forma, comprobaremos que esta aproximación es válida cuando se trabaja con valores de $\theta_0$ que se alejan de ser nulos.

<p style="text-indent: 2em; text-align: justify;"> Hemos de tener en cuenta también diferentes factores que afectan a que los resultados no sean los esperados según la teoría. Estamos haciendo suposiciones a la hora de construir estas fórmulas que no tienen por qué cumplirse en la realización del experimento. Hemos de considerar que la masa no es ideal y su momento de inercia no es el de una masa puntual, que el hilo tiene masa y está sujeto a tensiones variables que acompañan a la osiclación y que el péndulo no siempre oscila exactamente en un plano 2 dimensional. Todos estos factores contribuyen a aumentar el error en las mediciones realizadas.

### 7.1 Calculos de la experiencia 1

#### 7.1.1 Medidas directas

```python
def incertidumbreB_distribucion_triangular(resolucion):
    return resolucion / (24 ** 0.5)

def incertidumbreB_distribucion_rectangular(resolucion):
    return resolucion / (12 ** 0.5)

def incertidumbre_tipica_combinada(uA, uB):
    return (uA**2 + uB**2)**0.5

def incertidumbreA(s, N):
    return s / (N ** 0.5)
resolucion_cinta_metrica = 0.01 # cm
resolucion_cronometro    = 0.01 # s
```

```python
uBhi = incertidumbreB_distribucion_triangular(resolucion_cinta_metrica)
uBtj = incertidumbreB_distribucion_rectangular(resolucion_cronometro)
print(f"Incertidumbre B (h^i) : u_B(h^i) = {uBhi} cm")
print(f"Incertidumbre B (t^j) : u_B(t^j) = {uBtj} s")
```

```python
indices = np.array([1, 2, 3, 4, 5])
alturas_pendulo  = np.array([129.5, 110.0, 142.0, 161.5, 62.8])
numero_oscilaciones = np.array([20, 20, 20, 20, 20])
# t1, t2, t3: Medidas de tiempo en segundos para cada i
t1 = np.array([51.62, 52.58, 49.24, 45.72, 60.46])
t2 = np.array([50.81, 53.73, 48.40, 46.00, 60.63])
t3 = np.array([51.25, 54.08, 49.44, 45.70, 60.67])
```

```python
def formatear(lista, titulo="", unidad="s"): 
    print(titulo)
    for i in range(len(lista)):
        print(f"{i+1} : {lista[i]} {unidad}")
```

#### 7.1.2 Tiempo medio

Calculamos los tiempos medios para cada medida realizada: 

```python
t_medios = []
for i in range(len(indices)):
    tiempo_medio = (t1[i]+t2[i]+t3[i]) / 3
    t_medios.append(tiempo_medio)
formatear(t_medios, titulo="Tiempos medios <t_n>")
```

Calculamos las incertidumbres asociadas a los tiempos medios (tipo A, tipo B y combinada):

```python
# Incertidumbre tipo B
uB_t_medios = [uBhi, uBhi, uBhi, uBhi, uBhi]
# Incertidumbre tipo A
uA_t_medios = []
for i in range(len(indices)):
    s = np.std( [t1[i], t2[i], t3[i]] , ddof=1)
    uA_tiempo_medio = incertidumbreA(s, 3)
    uA_t_medios.append(uA_tiempo_medio)
# Incertidumbre típica combinada
u_tiempos_medios = []
for i in range(len(indices)):
    uA = uA_t_medios[i]
    uB = uB_t_medios[i]
    u_tiempo_medio = incertidumbre_tipica_combinada(uA, uB)
    u_tiempos_medios.append(u_tiempo_medio)
formatear(u_tiempos_medios, titulo="Incertidumbre tiempos medios u_<t_n>")
```

#### 7.1.3 Longitud

Calculamos las longitudes del péndulo para las cinco medidas tomadas: 

```python
H = 293 # cm 
longitudes_pendulo = [(H - h)/100 for h in alturas_pendulo] #Conversión a metros
formatear(longitudes_pendulo, titulo="Longitudes l", unidad="cm")
```

La incertidumbre de estas medidas será de tipo B, pues solo se ha tomado una medida de la altura para cada $1\leq i \leq 5$. Así:

```python
u_longitudes = []
for i in range(len(indices)):
    u_longitudes.append(uBhi)
formatear(u_longitudes, titulo="Incertidumbre longitudes u_l", unidad="cm")
```

#### 7.1.4 Periodo

Calculamos el valor medio del periodo (con su respectiva incertidumbre asociada):

```python
T_medios = [t_medios[i] / numero_oscilaciones[i] for i in range(len(indices))]
formatear(T_medios, titulo="Periodo T")
```

```python
u_periodos_medios = []
for i in range(len(indices)):
    u_periodos_medios.append(u_tiempos_medios[i] / numero_oscilaciones[i])
formatear(u_periodos_medios, titulo="Incertidumbre periodo medio u_T")
```

Ya en condiciones de calcular el periodo al cuadrado con su respectiva incertidumbre:

```python
T2_medios = [T**2 for T in T_medios]
formatear(T2_medios, titulo="Periodo cuadrado T^2", unidad = "s^2")
```

```python
u_periodos2_medios = []
for i in range(len(indices)):
    # Propagación de incertidumbres para T^2 = f(T) -> u(T^2) = g(T^2)
    u_periodo2_medio = T2_medios[i] *(2* (u_periodos_medios[i] / T_medios[i]) )
    u_periodos2_medios.append(u_periodo2_medio)
formatear(u_periodos2_medios, titulo="Incertidumbre periodo cuadrado medio u_T^2", unidad = "s^2")
```

#### 7.2 Ajuste lineal $T^2$ frente a $l$ 

```python
def Lineal (parameters, x):
    A, B = parameters
    y = np.multiply(A, x) + B
    return y 
lin_model = Model(Lineal)
data = RealData(longitudes_pendulo, T2_medios, sy = u_periodos2_medios)
odr = ODR(data, lin_model, beta0 = [0., 1.])
out = odr.run()
out.pprint()
```

```python
fit_y = Lineal (out.beta, longitudes_pendulo)
plt.errorbar(longitudes_pendulo, T2_medios, yerr = u_periodos2_medios, linestyle = "None", marker = "o", label = "Medidas")
plt.plot(longitudes_pendulo, fit_y, label = "Mínimos cuadrados")
plt.xlabel ("l (m)")
plt.ylabel ("T^2 (s^2)")
plt.grid()
plt.legend()
plt.show()
```

```python
# Incertidumbre asociada a a cada theta0: 
dy = H - h
uH = uBhi
uh = uBhi
udx = uBhi

u_theta0_exp2 = []
for dx in delta_x:
    denominador = dx**2 + dy**2
        
    termino_numerador = np.sqrt( (dx**2 * (uH**2 + uh**2)) + (dy**2 * udx**2) )
    u_theta_rad = (1 / denominador) * termino_numerador
        
    # 4. Conversión a grados
    theta_deg = np.degrees(theta0_exp2)
    u_theta_deg = np.degrees(u_theta_rad)
    u_theta0_exp2.append(u_theta_deg)

formatear(u_theta0_exp2, "Incertidumbre de los ángulos u_{theta_0}", "grados")
```

```python
sin2_exp2 = [np.sin(np.deg2rad(theta0 / 2))**2 for theta0 in theta0_exp2]
formatear(sin2_exp2, titulo="Seno cuadrado de theta_0 medios", unidad="")
```

```python

u_sin2_exp2 = []

for i in range(len(theta0_exp2)):
    th_rad = np.deg2rad(theta0_exp2[i]) # Ángulo en radianes    
    u_th_rad = np.deg2rad(u_theta0_exp2[i]) # Incertidumbre en radianes

    # Propagación
    incertidumbre = 0.5 * np.abs(np.sin(th_rad)) * u_th_rad
    
    u_sin2_exp2.append(incertidumbre)

formatear(u_sin2_exp2, titulo="Incertidumbre de sin^2(theta_0/2)", unidad="")
```
