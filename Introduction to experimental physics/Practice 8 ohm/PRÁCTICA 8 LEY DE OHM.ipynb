## 2. MARCO TEÓRICO GENERAL

<p style="text-indent: 2em; text-align: justify;"> Como se ha mencionado anteriormente, la ley de Ohm establece la relación:

$$ I = \frac{V}{R} \tag{1}$$

<p style="text-indent: 2em; text-align: justify;"> En general, los dispositivos eléctricos no poseen una resistencia constante, sino que es una función de la temperatura $T$. En algunos de ellos, el calentamiento provocado por la corriente aumenta la resistenciadel filamento. Por tanto, la ecuación ($1$) se convierte en:

$$ I (T) = \frac{V}{R(T)} \tag{2}$$

<p style="text-indent: 2em; text-align: justify;"> Si nos centramos ahora en otros dispositivos, como los diodos conductores, su resistencia depende de la dirección de la corriente. La relación intensidad-voltaje para un diodo semiconductor viene dada por la ecuación de Shockley:

$$ I = I_s \left(e^{\frac{V}{nV_o}} -1 \right) \tag{3}$$

<p style="text-indent: 2em; text-align: justify;"> donde $I_s \approx 10^{-12} - 10^{-6}$ es la corriente de saturación y $V_o = \frac{kT}{e}$, siendo $k$ la constante de Boltzmann, $T$ la temperatura absoluta y $e$ la carga del electrón. El parámetro $n$ depende del material que forma el diodo y toma valores típicos entre 1 para el germanio puro y 2 para el silicio puro.

### 4.3 ECUACIÓN DE SHOCKLEY

A partir de las medidas de voltaje en el diodo ($V_d$) y el voltaje en la resistencia de carga ($V_r$), se procedió a calcular la intensidad de corriente que atraviesa el dispositivo ($I_d$). Siguiendo la ley de Ohm y considerando la disposición del circuito de medida, la corriente se define como:

$$I_d = \frac{-V_r}{R_{nominal}}$$

donde $R_{nominal} = 1000 \, \Omega$. El signo negativo compensa la convención de medida para obtener valores de corriente positivos en polarización directa.

Para determinar el factor de idealidad ($n$) y la corriente de saturación inversa ($I_s$), se realizó un ajuste no lineal de los datos $(V_d, I_d)$ a la ecuación de Shockley (número 3).

$$I_d = I_s \left[ \exp\left( \frac{e \cdot V_d}{n \cdot k \cdot T} \right) - 1 \right]$$

Considerando una temperatura de laboratorio constante de $T = 297 \, K$. Debido a la sensibilidad de la función exponencial, el algoritmo solamente converge a los datos experimentales en la región de polarización directa. En la región de polaridad inversa, el voltaje medido es prácticamente nulo. Los parámetros optimizados obtenidos, con sus respectivas incertidumbres extraídas de la matriz de covarianza, son los siguientes:

$$n = 1.642 \pm 0.105$$
$$I_s = (2.07 \pm 0.35) \cdot 10^{-9} \, A$$

La incertidumbre asociada al factor $n$, $u_n$, se obtuvo mediante la raíz cuadrada del elemento correspondiente en la diagonal de la matriz de covarianza del ajuste:

$$u_n = \sqrt{var(n)}$$

## 5 CONCLUSIONES

### 5.3 Conclusiones sobre la experiencia del diodo

Al representar $I_d$ frente a $V_d$, se observa que las curvas no son simétricas. En polarización directa ($V > 0$), la corriente aumenta de forma exponencial, mientras que en polarización inversa ($V < 0$), la corriente es prácticamente nula. Físicamente, esto se debe a que el diodo actúa como una barrera de potencial unidireccional, permitiendo el flujo de corriente en un sentido y bloqueándola en el opuesto. Experimentalmente medimos que las tensiones del diodo en polarización inversa no eran exactamente cero. Aunque teóricamente el diodo es un elemento de resistencia infinita en una dirección de corriente, en la práctica esta "resistencia infinita" es inalcanzable.



Se ha comprobado que el comportamiento del diodo semiconductor se ajusta satisfactoriamente a la ecuación de Shockley. Además, la fuerte asimetría observada en la curva $I-V$ valida el carácter no óhmico del diodo.



El valor obtenido para el factor de idealidad, $n = 1.642$, resulta consistente con un diodo de silicio comercial. Al comparar este resultado con el valor teórico ideal de difusión pura ($n_{teórico} = 1$), se obtiene un error relativo porcentual del 64.2%:

$$\epsilon_r = \frac{n_{teórico} - n_{experimental}}{n_{teórico}} \times 100 = \frac{1 - 1.642}{1} \times 100 = -64.2\%$$

A pesar de la magnitud de este error respecto a la unidad, el valor es físicamente razonable. En los diodos de silicio reales, el factor $n$ suele oscilar en el rango $1 < n < 2$. Por otro lado, la corriente de saturación $I_s$ en el orden de los nanoamperios confirma que el material semiconductor es silicio, descartando el germanio, cuya $I_s$ sería órdenes de magnitud mayor.


Se confirma, por tanto,  el cumplimiento de los comportamientos predichos por las ecuaciones (1) a (3). La gráfica semilogarítmica muestra una tendencia lineal clara que valida la dependencia exponencial de la corriente respecto al voltaje.



A pesar del error con el valor teórico de $n$ para un diodo de silicio, la concordancia entre los datos experimentales y el ajuste teórico, especialmente visible en la escala semilogarítmica, sugiere que el error humano y las limitaciones del equipo de medida no han influido en exceso en esta práctica.

### 7.2 ECUACIÓN DE SHOCKLEY

```python
V_d = np.array([
    0.612, 0.607, 0.602, 0.594, 0.588, 0.578, 0.566, 0.548, 0.516, 0.449,
    -0.498, -0.986, -1.444, -1.938, -2.47, -2.93, -3.46, -3.95, -4.50, -4.93
])

V_r = np.array([
    -4.34, -3.86, -3.42, -2.85, -2.40, -1.921, -1.435, -0.900, -0.406, -0.570,
    0.0, 0.01e-3, 0.01e-3, 0.02e-3, 0.0002, 0.0003, 0.0003, 0.0004, 0.0004, 0.0005
])

R_nominal = 1000 # R nominal coincide con la R medida

T_ambiente = 297 


```

```python
def corriente_diodo_shockley(V, Is, n):
    k = 1.381e-23 
    e = 1.602e-19 
    V0 = (k * T_ambiente) / e
    exponente = np.clip(V / (n * V0), None, 700) 
    return Is * (np.exp(exponente) - 1)
```

```python

I_diodo = -V_r / R_nominal 
# Ajusto por mínimos cuadrados no lineales cuando la polaridad es directa porque si no
# no puedo calcular la matriz de covarianza y se queda "pillado" en el cero (porque en
# polaridad inversa, el voltaje es prácticamente cero y entonces scipy ajusta a
# f(Vd) = "Id"(Vd) = 0 )
mascara_directa = V_d > 0.4
V_fit = V_d[mascara_directa]
I_fit = I_diodo[mascara_directa]


parametros_optimos, matriz_covarianza = curve_fit(
    corriente_diodo_shockley, 
    V_fit, 
    I_fit, 
    p0=[1e-12, 1.5],
    bounds=([1e-15, 1.0], [1e-7, 2.5])
)

i_s, n = parametros_optimos
incertidumbres = np.sqrt(np.diag(matriz_covarianza))
un = incertidumbres[1]

print(f"n: {n:.3f} ± {un:.3f}")
print(f"Is: {i_s:.3e} A")
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(V_d, I_diodo, color='red', label='Datos experimentales')
voltaje_continuo_teorico = np.linspace(min(V_d), max(V_d), 500)
corriente_ajustada_teorica = corriente_diodo_shockley(voltaje_continuo_teorico, *parametros_optimos)

plt.plot(
    voltaje_continuo_teorico, 
    corriente_ajustada_teorica, 
    linestyle='--', 
    color='blue', 
    label=f'Ajuste Shockley (n={n:.2f})'
)

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.xlabel('Voltaje en el diodo ($V_d$) (V)')
plt.ylabel('Corriente en el diodo ($I_d$) (A)')
plt.title('Curva I-V para el Diodo ')
plt.legend()
plt.grid(True, alpha=0.3)
plt.ylim(min(I_diodo)-0.0005, max(I_diodo)+0.0005) 
plt.show()
```
