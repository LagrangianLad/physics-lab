## 1. INTRODUCCIÓN Y OBJETIVOS

<p style="text-indent: 2em; text-align: justify;"> La mecánica de fluidos es la rama de la física que se dedica a estudiar fluidos así como su comportamiento bajo diferentes condiciones. Ello nos permite someterlos a cambios de presión, temperatura, altura y observar cómo varía su flujo a través de un contenedor, que normalmente es un tubo. Así, somos capaces de extraer propiedades intrínsecas a cada fluido, como lo pueden ser su viscosidad, es decir, cómo se adhiere a las paredes del recipiente que lo contiene, o su densidad, siendo esta su relación entre su masa y volumen.

<p style="text-indent: 2em; text-align: justify;"> En particular, es curioso hacer un estudio de la diferencia existente cuando el fluido se encuentra en los conocidos como régimen laminar y régimen turbulento. El número de Reynolds nos indica en el régimen en el que está el fluido. Si es menor que 2300 aproximadamente, el régimen es laminar y, si es mayor a 4000, turbulento. Como dicho número es directamente proporcional a la velocidad del fluido (e inversamente proporcional a la viscosidad de este), en este experimento se trabajará con velocidades del fluido no muy elevadas para trabajar siempre en régimen laminar. De esta forma, serán eficaces las aproximaciones de primer orden realizadas en los cálculos teóricos.

<p style="text-indent: 2em; text-align: justify;"> Por tanto, esta práctica tiene como objetivos principales comprobar la validez de la Ley de Poiseuille y calcular experimentalmente el valor de la viscosidad del agua.

## 2. MARCO TEÓRICO GENERAL

<p style="text-indent: 2em; text-align: justify;"> Consideremos un tubo de igual sección en los extremos, por el que circula un fluido, con sus extremos A y B a la misma altura. Debido a efectos de viscosidad, entre los putnos A y B se crea una diferencia de presión $\Delta P$. El caudal que circula por el tubo depende del volumen que pasa por su sección por unidad de tiempo y viene dado por la expresión:

$$ Q = \frac{V}{t} \tag{1}$$

<p style="text-indent: 2em; text-align: justify;"> En caso de régimen laminar, la Ley de Poiseuille establece la siguiente relación:

$$ \Delta P = P_A - P_B = RQ \tag{2}$$

<p style="text-indent: 2em; text-align: justify;"> Siendo R la resistencia al flujo que ejerce el tubo. Si suponemos el tubo de sección constante y circular, podemos explicitar su valor mediante la expresión:

$$ R = \frac{8 \eta L}{\pi r^4} \tag{3}$$

<p style="text-indent: 2em; text-align: justify;">Donde $\eta $ es el coeficiente de viscosidad del fluido, $L$ la longitud del tubo y $r$ el radio interno del mismo. Introduciendo esta expresión en ($2$):

$$ Q = \frac{\pi r^4}{8 \eta L} \Delta P \tag{4}$$

<p style="text-indent: 2em; text-align: justify;">De esta forma, podremos hacer un ajuste por mínimos cuadrados de $Q$ frante a $\Delta P$ a fin de obtener el valor de $\eta$ de manera experimental.

<p style="text-indent: 2em; text-align: justify;"> Se tendrá en cuenta que el valor real de la viscosidad del agua es de $0.9 \cdot 10^{-3} Pa.s$ a una temperatura de 25 ºC

### 4.2 REGRESIÓN LINEAL

$$ u_{m_{agua}} = \sqrt{2}\cdot u_{balanza}$$

donde asumimos que la balanza sigue una distribución de probabilidad triangular y tiene asociada por tanto la siguiente incertidumbre:

$$ u_{balanza} = \frac{\Delta}{\sqrt{12}}$$

Una vez obtenida esta incertidumbre, procedemos al cálculo de la del volumen. Mediante las derivadas parciales, obtenemos:

$$ u_V = \frac{\partial V}{\partial m} \cdot u_{m_{agua}} = \frac{1}{\rho_{agua}} \cdot u_{m_{agua}}$$

Por último, para hallar la incertidumbre asociada al cálculo final del caudal, volvemos a aplicar la propagación. El cálculo arroja:

$$ u_Q = Q \cdot \sqrt{\left(\frac{u_V}{V}\right)^2 +\left(\frac{u_t}{t}\right)^2}$$

Procedemos por tanto a realizar la regresión lineal, de la cual obtenemos la siguiente gráfica con la pendiente asociada:

$$ \eta = \frac{\pi r^4}{8 a L}$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.odr import *
```

```python
rho_agua = 1 # g/cm^3
g = 9.80 # m/s^2
rho_agua_SI = 1000 # kg/m^3
```

```python
l_tubo = 39.5 #cm
l_a = [14.5, 15.4, 18.6, 19.8, 26.2, 23.3, 21.5, 27.7, 36.7, 41.3] # cm
l_b = [14.9, 15.7, 20.3, 21.7, 28.3, 24.7, 22.6, 30.0, 39.8, 45.3] # cm
d_h = [l_b[i] - l_a[i] for i in range(len(l_a))] # cm
d_h_m = [elemento/100 for elemento in d_h] # m

t = [120.13, 119.98, 120.10, 119.56, 120.13, 120.39, 120.06, 60.03, 60.07, 60.32] # s

m = [48.66, 56.13, 89.27, 92.78, 131.24, 104.11, 89.15, 79.68, 93.60, 103.98] # g
m_vaso = 19.55 #g
m_agua = [(m[i] - m_vaso)*0.001 for i in range(len(m))]

V_agua_SI = [m_agua[i]/rho_agua_SI for i in range(len(m))] # m^3

d_P = [rho_agua_SI*g*altura for altura in d_h_m] # Pa

Q = [V_agua_SI[i]/t[i] for i in range (len(t))]
```

```python
# INCERTIDUMBRES

def uB_digital(res):
    return res/np.sqrt(12)

def uB (res):
    return res/np.sqrt(24)

def uA (s, N):
    return s/np.sqrt(N)

def u_combinada (u_a, u_b):
    return np.sqrt(pow(u_a, 2)+ pow(u_b, 2))

def imprimir (lista, unidades, titulo = ""):
    print(titulo)
    for i in range(len(lista)):
        print(f"{i+1} : {lista[i]} {unidades}")

u_balanza = uB_digital(0.01)
u_metro = uB(0.1)
u_tiempos = uB_digital (0.01)

u_m_agua = (np.sqrt(2)*u_balanza)*0.001 #kg
u_V_agua_SI = u_m_agua/rho_agua_SI

u_Q = [Q[i]*np.sqrt(pow((u_V_agua_SI)/V_agua_SI[i], 2)+ pow((u_tiempos)/t[i], 2)) for i in range(len(t))]
```

```python
# IMPRIMIR

imprimir(Q, r"$m^3/s$", "Caudal")
imprimir(u_Q, r"$m^3/s$", "Incertidumbre Caudal")
```

```python
# REGRESIÓN LINEAL

def Lineal (parameters, x):
    A, B = parameters
    y = np.multiply(A, x) +B
    return y 

lin_model = Model(Lineal)
data = RealData(d_P, Q, sy = u_Q)
odr = ODR(data, lin_model, beta0 = [0.,1.])
out = odr.run()
out.pprint()
```

```python
fit_y = Lineal(out.beta, d_P)
plt.errorbar(d_P, Q, yerr= u_Q, linestyle = "None", marker = 'o', label = r"$u_Q$")
plt.plot(d_P, fit_y, label = r"$\Delta P$ frente a Q")
plt.xlabel (r"$\Delta P$ (Pa)")
plt.ylabel(r"Q ($m^3/s$)")
plt.legend()
plt.show()
```
