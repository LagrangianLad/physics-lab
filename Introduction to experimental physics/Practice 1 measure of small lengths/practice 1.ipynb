# 2. PROCEDIMIENTO Y MONTAJE EXPERIMENTAL

<p style="text-indent: 2em; text-align: justify;"> Como se ha mencionado anteriormente, el objeto de esta práctica será medir las longitudes y masas de los susodichos tres objetos así como de las arandelas con el fin de determinar sus densidades. Para la medida de las dimensiones del el cilindro, el paralelepípedo y la placa usaremos un pie de rey. Al tratarse de objetos cuyas dimensiones no se alejan en gran cantidad de la resolución del instrumento, las medidas tomadas serán fidedignas en su mayoría. Así pues, empleamos dicho aparato correctamente y recabamos la información sobre las dimensiones de los objetos para anotarlas en tablas de resultados posteriormente. Análogamente, haremos uso de una balanza digital para determinar sus masas. Una vez disponemos de toda esta información, utilizaremos las siguientes fórmulas matemáticas para hallar la densidad.

<p style="text-indent: 2em; text-align: justify;"> Para el cilindro, sean $h$ su altura, $d$ su diámetro; para la placa y el paralelepípedo, sean $x$ su ancho, $y$ su largo y $z$ su alto. Entonces: $$\begin{cases} V_{cilindro} = \frac{\pi}{4}h\cdot d^2\\ V_{placa} = x\cdot y \cdot z\\V_{paralelepípedo} = x  \cdot y \cdot z\end{cases}$$

<p style="text-indent: 2em; text-align: justify;"> Sabido esto, y si suponemos que la distribución de masas en los materiales es homogénea, la fórmula para el cálculo de densidad es común a todas las figuras:

$$
\rho = \frac{m}{V} \tag{1}
$$
<p style="text-indent: 2em; text-align: justify;"> Con respecto a las arandelas, para medir sus radios tanto interior como exterior volveremos a hacer uso del pie de rey. No obstante, para medir el espesor recurriremos al micrómetro, herramienta mucho más precisa que el pie de rey, indicada para la medición de longitudes pequeñas. Anotaremos el espesor tanto de una arandela como de 10 de ellas. Para la masa, recabaremos, con la ayuda de la balanza, la masa de 0, 5, 10, 15, 20, 25, 30 arandelas. Posteriormente, se estimará la masa por regresión lineal, considerando la ecuación $$ m_n = m_1\cdot n \tag{2}$$

<p style="text-indent: 2em; text-align: justify;">donde n es el número de arandelas y m la masa en gramos. La pendiente de dicha recta será la masa de una arandela, que es el dato buscado.

<p text-align: justify;">No obstante, será preciso tener en cuenta la propagación de incertidumbres a la hora de calcular tanto el volumen como la densidad. La incertidumbre asociada al cálculo del volumen del paralelepípedo y la placa es

$$\left(\frac{u_V}{\overline{V}}\right)^2 = \sum_{k=x,y,z}\left(\frac{u_k}{\overline{k}}\right)^2$$

<p text-align: justify;">Si nos referimos al volumen del cilindro:

$$ {\left(\frac{u_V}{V}\right)}^2 = {\left(2\frac{u_d}{d}\right)}^2 + {\left(\frac{u_h}{h}\right)}^2 \tag{4}$$
<p text-align: justify;">En cuanto a la densidad de cualquiera de las figuras:

$$\left( \frac{u_{\rho}}{\rho} \right)^2 = \left( \frac{u_m}{m} \right)^2 + \left( \frac{u_V}{V} \right)^2 \tag{5}$$

<p text-align: justify;">Además, la incertidumbre total será la el cuadrado de la suma de las incertidumbres tipo a (debidas a la imposibilidad de medir siempre los mismos valores) y las incertidumbres tipo b (derivadas de la resolución de los instrumentos usados, mayoritariamente). Así:

$$ u^2 = u_A^2 + u_B^2 \tag{6}$$

#### Cálculo del volumen de una arandela por Método I y Método II

El volumen de una arandela se determina mediante la fórmula geométrica de un cilindro hueco:
$$V_1 = \frac{\pi}{4} \cdot (d_{exterior}^2 - d_{interior}^2) \cdot e_1$$
Donde V denota el volumen, $d$ el diámetro (exterior e interior) y $e$ el espesor del cilindro. En general reservaremos el subíndice $n\in\mathbb{N}$ para distinguir propiedades asociadas a $n$ arandelas. Así $V_n$ será el volumen de $n$ arandelas apiladas y $e_n$ el espesor de $n$ arandelas apiladas.

##### 1. Método I: Medida de una arandela individual
En este método se miden directamente las dimensiones de una sola pieza. Al realizarse una única toma de datos, la incertidumbre de tipo A es cero, por lo que solo se considera la incertidumbre de tipo B asociada a la resolución de los instrumentos.

Tras aplicar la fórmula de propagación de errores concluimos que: 
$$V_1 = \overline{V_1} \pm u_{V_1} = 993.3160 \pm 0.0096 \text{ mm}^3$$

_Véase el anexo 6.2.A_
##### 2. Método II: Medida de diez arandelas

Este método consiste en medir el espesor total de un bloque de $n=10$ arandelas y posteriormente dividir los resultados entre diez para obtener el volumen de una arandela.


$$V_1 = \overline{V_1} \pm u_{V_1} = 1183.364 \pm 0.011 \text{ mm}^3$$
_Véase el anexo 6.2.B_


>  Existe una diferencia notable entre ambos métodos. El Método II arroja un volumen mayor, lo que podría deberse a irregularidades en la superficie de las arandelas que impiden un ajuste perfecto al apilarlas, sumando aire o "espacios muertos" al espesor total medido ($e_{10}$). 


#### Cálculo de la masa de una arandela


Para determinar la masa promedio de una arandela, se ha realizado una regresión lineal relacionando el número de arandelas con la masa total medida.

La masa promedio de una arandela individual corresponde directamente a la pendiente de la recta de regresión. El error de cero forzado (ordenada en el origen) es prácticamente despreciable frente al rango de masas y no influye en el cálculo de la masa unitaria.

El valor de la masa promedio por arandela, expresado con su incertidumbre asociada, es:

$$m_1 = \overline{m_1} \pm u_{m_1} = 7.808 \pm 0.040 \text{ g}$$

#### Determinación de la densidad de una arandela

Como dependiendo del método elegido para medir el volumen, hemos obtenido una medida de $V_1$ diferente, vamos a especificar las densidades medidas según ambos métodos. La densidad promedio medida será:

$$\overline{\rho_I} = \frac{\overline{m_1}}{\overline{V_{1_{I}}}} = 7.861\frac{\text{g}}{\text{cm}^3}$$
para el método I, así como: 
$$\overline{\rho_{II}} = \frac{\overline{m_2}}{\overline{V_{1_{II}}}} =  6.598\frac{\text{g}}{\text{cm}^3}$$
para el método II. 

Además podemos calcular las incertidumbres asociadas a estas medidas por la fórmula de propagación de incertidumbres:

$$u_{\rho_k} = \overline{\rho_k}\sqrt{\left(\frac{u_m}{\overline{m}}\right)^2 + \left(\frac{u_{V_k}}{\overline{V_{k}}}\right)^2} \text{    ;   } k = I, II$$

Para el método $I$ tendremos:

$$u_{\rho_I} = 0.007861\cdot \sqrt{\left(\frac{0.040}{7.808}\right)^2 + \left(\frac{0.0096}{993.3160}\right)^2} = 0.0000040 \frac{\text{g}}{\text{mm}^3} = 0.040 \frac{\text{g}}{\text{cm}^3}$$

Y para el método $II$: 

$$u_{\rho_{II}} = 0.006598\cdot \sqrt{\left(\frac{0.040}{7.808}\right)^2 + \left(\frac{0.011}{1183.364}\right)^2} = 0.000030 \frac{\text{g}}{\text{mm}^3} = 0.034 \frac{\text{g}}{\text{cm}^3}$$

Así podremos finalmente expresar las densidades como:

$$\boxed{
\rho_I = \overline{\rho_I} \pm u_{\rho_I} = 7.861 \pm 0.040 \frac{\text{g}}{\text{cm}^3} 
\text{     ;     } \rho_{II} = \overline{\rho_{II}} \pm u_{\rho_{II}} = 6.598 \pm 0.034 \frac{\text{g}}{\text{cm}^3} }$$

### EXPERIMENTO 1

Los valores medios para las densidades de los objetos medidos son:

- Cilindro: $2.809 \frac{g}{{cm}^3}$
- Paralelepípedo: $2.670 \frac{g}{{cm}^3}$ 
- Placa delgada metálica: $2.720 \frac{g}{{cm}^3}$

El valor bibliográfico de la densidad del aluminio es de aproximadamente $2.698 \frac{g}{{cm}^3}$. Lo que nos lleva a pensar que las tres piezas son de este material. Este hecho es inesperado pues a simple vista intentamos escoger la placa delgada de un metal diferente. Podemos proceder al cálculo del error relativo cometido en la obtención de la densidad de cada pieza: 

$$ \epsilon_{relativo} = {\frac{\rho_{experimental}-\rho_{bibliográfica}}{\rho_{bibliográfica}}} $$

-Para el cilindro: $\epsilon_{relativo} = \frac{2.809-2.698}{2.698} = 0.041$

-Para el paralelepípedo: $\epsilon_{relativo} = \frac{2.670 - 2.698}{2.698} = -0.010$

-Para la placa delgada: $\epsilon_{relativo} = \frac{2.720-2.698}{2.698} = 0.008$

Estos resultados nos indican que las piezas podrían ser de aluminio perfectamente ya que el error cometido al comparar la densidad obtenida con la bibliográfica es menor o igual al 4% en todos los casos. Por tanto, podemos concluir lo siguiente:
    
1. Los instrumentos utilizados en este experimento proporcionan medidas acertadas y son adecuados para determinar las densidades de objetos de dimensiones similares a los utilizados.
2. La densidad obtenida con más precisión ha sido la de la placa delgada, con un error relativo cercano al 0.8%. El cilindro ha sido la pieza que mayor error presenta. Lo cual nos lleva a que la instrumentación empleada en el experimento aporta buenos resultados cuando se manejan objetos de dimensiones y forma similares a la placa, y no al cilindro, ya que, por ejemplo, no resulta sencillo alinear el pie de rey para que pase exactamente por el centro de la base del cilindro a la hora de calcular su diámetro.

<p style="text-indent: 2em; text-align: justify;">Un punto central de análisis es la comparación de la incertidumbre al medir el espesor de una sola arandela frente a un bloque de diez. <p style="text-indent: 2em; text-align: justify;">Para la determinación de la masa, se evitó la medida directa de una única unidad, optando en su lugar por un ajuste de regresión lineal. La pendiente de la recta resultante proporcionó una estimación de la masa unitaria. Se indujo a la hora de hacer el experimento un error de cero en la balanza, de tal forma que esta no marcara $0.00 g$ cuando no tenía ninguna arandela depositada sobre ella. No obstante, este error de cero no afecta a la determinación de $m_1$ mediante el ajuste por mínimos cuadrados ya que la pendiente de una recta es independiente de su ordenada en el origen. Por tanto, el error de cero de la balanza no afecta a la determinación de $m_1$

<p style="text-indent: 2em; text-align: justify;">Finalmente, el cálculo de las densidades arroja dos valores diferentes: 
$$\rho_I = \overline{\rho_I} \pm u_{\rho_I} = 7.861 \pm 0.040 \frac{\text{g}}{\text{cm}^3} 
\text{     ;     } \rho_{II} = \overline{\rho_{II}} \pm u_{\rho_{II}} = 6.598 \pm 0.034 \frac{\text{g}}{\text{cm}^3} $$    
<p style="text-indent: 2em; text-align: justify;">Esta discrepancia señala que la precisión no depende únicamente del instrumento utilizado, sino de la aplicación de métodos indirectos como el apilamiento de las arandelas.

<p style="text-indent: 2em; text-align: justify;">Si consultamos la densidad biliográfica del acero, apreciamos que es de $7.850 \frac{g}{cm^3}$. Por tanto, las arandelas podrían estar hechas de ese material según los resultados experimentales. Aunque la densidad de las arandelas medidas con el segundo método sea mucho menor, esto cuadra con la explicación de que la curvatura de las arandelas imposibilita modelar su volumen como un cilindro semi hueco sólido y homogéneo. En dicho caso, podemos obtener el error relativo de forma análoga al experimento 1:

- $\epsilon_1 = \frac{7.850-7.861}{7.850} = -0.001401$
- $\epsilon_2 = \frac{7.850-6.598}{7.850} = 0.15949$

Podemos entonces calcular los valores medios de las medidas tomadas: 

$$\overline{h} = \frac{1}{5}\sum_{i=1}^5h_i = \frac{21.60+ 21.60+21.60+ 21.60+21.60}{5} = 21.60 \text{ mm}$$
$$\overline{d} = \frac{1}{5}\sum_{i=1}^5d_i = \frac{25.15+ 25.25+ 25.25+ 25.15+ 25.25}{5} = 25.21 \text{ mm}$$
$$\overline{m} = \frac{1}{5}\sum_{i=1}^5m_i = \frac{30.287+ 30.286+ 30.284+ 30.286+ 30.285}{5} = 30.2856 \text{ g}$$


Así como las varianzas muestrales de las tres magnitudes

$$s_h^2 = \frac{1}{5-1}\sum_{i=1}^5(h_i - \overline{h})^2 = \frac{4\cdot(21.60 - 21.60)}{4} = 0.000 \text{ mm}^2$$
$$s_d^2 = \frac{1}{5-1} \sum_{i=1}^{5} (d_i - \overline{d})^2 = \frac{(25.15-25.21)^2 + 3(25.25-25.21)^2 + (25.15-25.21)^2}{4} = 0.003 \text{ mm}^2$$
$$s_m^2 = \frac{1}{5-1} \sum_{i=1}^{5} (m_i - \overline{m})^2 = \frac{(30.287-30.2856)^2 + (30.286-30.2856)^2 + (30.284-30.2856)^2 + (30.286-30.2856)^2 + (30.285-30.2856)^2}{4}=0.0000013 \text{g}^2$$

```python
altura_media_cilindro = np.mean(alturas_cilindro)       # mm
diametro_medio_cilindro = np.mean(diametros_cilindro)   # mm
masa_media_cilindro = np.mean(masas_cilindro)           # g
print(f" Altura media cilindro = {altura_media_cilindro}  mm\n Diámetro medio cilindro = {diametro_medio_cilindro}  mm\n Masa media cilindro = {masa_media_cilindro}  g")
```

```python
altura_varianza_cilindro = np.var(alturas_cilindro, ddof=1)     # mm^2
diametro_varianza_cilindro = np.var(diametros_cilindro, ddof=1) # mm^2
masa_varianza_cilindro = np.var(masas_cilindro, ddof=1)         # g^2
print(f" Altura varianza = {round(altura_varianza_cilindro, 7)} mm^2\n Diámetro varianza = {round(diametro_varianza_cilindro,7)}  mm^2\n Masa varianza = {round(masa_varianza_cilindro,7)} g^2")
```

Así podemos calcular la incertidumbre tipo A de cada una de las magnitudes medidas (expresadas con dos cifras significativas)

$$u_A(h) = \frac{s_h}{5} = \frac{\sqrt{0}}{\sqrt{5}} = 0 \text{ mm} \implies u_A(h)^2 = 0 \text{ mm}^2$$
$$u_A(d) = \frac{s_d}{5} = \frac{\sqrt{0.003}}{\sqrt{5}} = 0.024 \text{ mm}\implies u_A(d)^2 = 0.00060 \text{ mm}^2$$
$$u_A(m) = \frac{s_m}{5} = \frac{\sqrt{0.0000013}}{\sqrt{5}} = 0.00051 \text{ g}\implies u_A(m)^2 = 0.00000026 \text{ g}^2$$

```python
def uA2(s2x, N=5): # u tipo A cuadrado
    return ((s2x / N)**0.5)**2

uA2_altura_cilindro = uA2(altura_varianza_cilindro)      # mm^2
uA2_diametro_cilindro = uA2(diametro_varianza_cilindro)  # mm^2
uA2_masa_cilindro = uA2(masa_varianza_cilindro)          # g^2

print(f" INCERTIDUMBRE TIPO A CUADRADO\n uA2 altura = {uA2_altura_cilindro} mm^2\n uA2 diámetro = {round(uA2_diametro_cilindro, 7)}  mm^2\n uA2 masa = {round(uA2_masa_cilindro,8)} g^2")
```

Para calcular la incertidumbre de tipo B de la altura y el diámetro asumiremos que la medida del pie de rey viene gobernada por una distribución triangular de probabilidad. La expresaremos con dos cifras significativas, análogamente a la incertidumbre tipo A.
$$u_B(h) = u_B(d) = \frac{\Delta}{\sqrt{24}} = \frac{0.05}{\sqrt{24}} = 0.010 \text{ mm}\implies u_B(h)^2 = u_B(d)^2 = 0.00010 \text{ mm}^2$$

```python
def uB2_triangular(resolucion): # u tipo B cuadrado (distribución de probabilidad triangular)
    return (resolucion / (24)**0.5)**2
```

Para calcular la incertidumbre de tipo B de la masa (medida con una balanza digital de resolución $\Delta = 0.001 \text{g}$) asumiremos que la medida viene gobernada por una distribución unifore de probabilidad. Entonces:

$$u_B(m) = \frac{\Delta}{\sqrt{12}} = 0.00029 \implies u_B(m)^2 = 0.000000083 \text{ g}^2$$

```python
def uB2_uniforme(resolucion): # u tipo B cuadrado (distribución de probabilidad uniforme)
    return (resolucion / (12)**0.5)**2

res_pie_rey = 0.05   # mm
res_balanza = 0.001  # g

uB2_altura_cilindro = uB2_triangular(res_pie_rey)    # mm^2
uB2_diametro_cilindro = uB2_triangular(res_pie_rey)  # mm^2
uB2_masa_cilindro = uB2_uniforme(res_balanza)        # g^2

print(f" INCERTIDUMBRE TIPO B CUADRADO\n uB2 altura = {round(uB2_altura_cilindro, 4)} mm^2\n uB2 diámetro = {round(uB2_diametro_cilindro,4)}  mm^2\n uB2 masa ={round(uB2_masa_cilindro,9)} g^2")
```

Así podemos calcular las incertidumbres típicas combinadas de las tres medidas:

$$u_h^2 = u_A(h)^2 + u_B(h)^2 \implies u_h = \sqrt{0 + 0.00010} = 0.010 \text{ mm}$$
$$u_d^2 = u_A(d)^2 + u_B(d)^2 \implies u_d = \sqrt{0.00058 + 0.00010} = 0.026 \text{ mm}$$
$$u_m^2 = u_A(m)^2 + u_B(m)^2 \implies u_m = \sqrt{0.00000026 + 0.000000084} = 0.00059 \text{ g}$$


```python
def u(uA2, uB2): # Incertidumbre tipica combinada
    return (uA2 + uB2)**0.5

u_altura_cilindro = u(uA2_altura_cilindro, uB2_altura_cilindro)         # mm
u_diametro_cilindro = u(uA2_diametro_cilindro, uB2_diametro_cilindro)   # mm
u_masa_cilindro = u(uA2_masa_cilindro, uB2_masa_cilindro)               # g

print(f"INCERTIDUMBRES TÍPICAS COMBINADAS\n u altura = {round(u_altura_cilindro,4)} mm \n u diámetro = {round(u_diametro_cilindro,4)}  mm \n u masa = {round(u_masa_cilindro,5)} g")
```

Que nos permite expresar la medida de las magnitudes como:

$$h = \overline{h} \pm u_h = 21.600 \pm 0.010 \text{ mm}$$
$$d = \overline{d} \pm u_d = 25.210 \pm 0.026 \text{ mm}$$
$$m = \overline{m} \pm u_m = 30.28560 \pm 0.00059 \text{ g}$$

```python
volumen_medio_cilindro = 10781.7516  # mm^3
u_volumen_cilindro = volumen_medio_cilindro * ((u_altura_cilindro / altura_media_cilindro)**2 + (2*(u_diametro_cilindro / diametro_medio_cilindro))**2)**0.5 # mm^3
print(f"Volumen del cilindro = {volumen_medio_cilindro} ± {round(u_volumen_cilindro,4)} mm^3")
```

```python
densidad_media_cilindro = masa_media_cilindro / volumen_medio_cilindro # g / mm^3
u_densidad_cilindro = densidad_media_cilindro * ((u_masa_cilindro / masa_media_cilindro)**2 + ((u_volumen_cilindro / volumen_medio_cilindro))**2)**0.5 # g/mm^3
print(f"Densidad del cilindro = {round(densidad_media_cilindro,8)} ± {round(u_densidad_cilindro,7)} g / mm^3")
```

#### 6.2.A Cálculo del volumen por el método $I$: medida de una arandela

Comencemos por determinar el volumen de una arandela así como su incertidumbre. Como solo hemos tomado una medida del radio exterior, interior y del espesor de una arandela, el valor promedio de estas magnitudes coincide con el valor medido. Así, un simple cálculo geométrico nos confirma que:

$$V = \frac{\pi}{4}(d_{exterior}^2 - d_{interior}^2)\cdot e_1 \implies \overline{V_1} = \frac{\pi}{4} \cdot (28.40^2 - 12.70^2) \cdot 1.96 = 993.3160 \text{ mm}^3$$

La incertidumbre de tipo A del volumen medido de esta forma será cero (porque todas las varianzas valen cero, al haber únicamente una medida, esta coincide con el promedio de cada magnitud). Por otra parte podemos calcular la incertidumbre tipo B de cada magnitud. Si suponemos que tanto el pie de rey como el micrómetro siguen una distribución de probabilidad triangular, entonces:

$$u_B(r_{exterior}) = u_B(r_{interior}) = 0.00010 \text{ mm} \text{  ;  } u_B(e_1) = 0.000000042 \text{ mm}$$

La incertidumbre del volumen vendrá entonces dada por: 


$$u_V = \sqrt{\left( \frac{\partial V}{\partial d_{exterior}} \right)^2 u_B^2(d_{exterior}) + \left( \frac{\partial V}{\partial d_{interior}} \right)^2 u_B^2(d_{interior}) + \left( \frac{\partial V}{\partial e} \right)^2 u_B^2(e)}$$

Calculo las derivadas parciales que aparecen en la propagación: 
 $$\frac{\partial V}{\partial d_{exterior}} = \frac{\pi}{2} \cdot d_{exterior} \cdot e = \frac{\pi}{2} \cdot 28.40 \cdot 1.96 = 87.44 \text{ mm}$$
 $$\frac{\partial V}{\partial d_{interior}} = \frac{-\pi}{2}\cdot d_{interior} \cdot e = \frac{-\pi}{2} \cdot 12.70 \cdot 1.96 = -39.10 \text{ mm}$$
 $$\frac{\partial V}{\partial e} = \frac{\pi}{4} \cdot (r_{ext}^2 - r_{int}^2) = \frac{pi}{4} \cdot (28.40^2 - 12.70^2) = 506.79 \text{ mm}$$

Entonces, sustituyendo (teniendo en cuenta que al solo haber tomado una medida, la varianza muestral es cero y por tanto la incertidumbre tipo A es cero): 
$$u_V = \sqrt{(87.44)^2 \cdot (0.00010)^2 + (-39.10)^2 \cdot (0.00010)^2 + (506.79)^2 \cdot (0.000000042)^2} = 0.0096 \text{ mm}^3$$


El volumen de una arandela individual, expresado con su incertidumbre es:

$$V_1 = \overline{V_1} \pm u_{V_1} = 993.3160 \pm 0.0096 \text{ mm}^3$$

#### 6.2.B CÁLCULO DEL VOLUMEN POR EL MÉTODO II: MEDIDA DE DIEZ ARANDELAS

El espesor determinado experimentalmente para diez arandelas es: 
$$\overline{e_{10}} = 23.35 \text{ mm}$$
y por tanto el volumen de una arandela promedio será:
$$\overline{V_1} = \frac{\pi}{4} \cdot (d_{exterior}^2 - d_{interior}^2)\cdot \frac{\overline{e_{10}}}{10}=\frac{\pi}{4}\cdot(28.40^2 - 12.70^2)\cdot2.335 = 1183.364 \text{ mm}^3$$
Al haber tomado únicamente una medida experimental de este espesor, su incertidumbre tipo A asociada es cero. La incertidumbre tipo B asociada coincidirá con la incertidumbre tipo B para el espesor calculada en el anexo 6.2.A
$$u_B(e_{10}) = 0.000000042 \text{ mm}$$
Puesto que los instrumentos de medición empleados han sido los mismos. 
Al ser esta la incertidumbre asociada al espesor de diez arandelas, la incertidumbre asociada al espesor de una única arandela será:$$u_B(e_{1}) = \frac{u_B(e_10)}{10} = 0.0000000042 \text{ mm}$$


La incertidumbre del volumen vendrá entonces dada por la misma fórmula de propagación que en el anexo anterior (6.2.1): 


$$u_{V_1}= \sqrt{\left( \frac{\partial V}{\partial d_{exterior}} \right)^2 u_B^2(d_{exterior}) + \left( \frac{\partial V}{\partial r_{interior}} \right)^2 u_B^2(d_{interior}) + \left( \frac{\partial V}{\partial e} \right)^2 u_B^2(e)}$$

Si además asumimos que los radios exteriores e interiores coinciden para todas las arandelas, y que el espesor de una arandela es la décima parte del espesor de las diez:
 $$\frac{\partial V}{\partial d_{exterior}} = \frac{\pi}{2} \cdot d_{exterior} \cdot e_1 = \frac{\pi}{2} \cdot 28.40 \cdot 2.335 = 104.16 \text{ mm}$$
 $$\frac{\partial V}{\partial d_{interior}} = \frac{-\pi}{2} \cdot d_{int} \cdot e = \frac{-\pi}{2} \cdot 12.70 \cdot 2.335 = 46.58 \text{ mm}$$
 $$\frac{\partial V}{\partial e} = \frac{\pi}{4} \cdot (d_{ext}^2 - d_{int}^2) = \frac{\pi}{4} \cdot (28.40^2 - 12.70^2) = 2027.17 \text{ mm}$$

$$u_{V_1} = \sqrt{(104.16)^2 \cdot (0.00010)^2 + (46.58)^2 \cdot (0.00010)^2 + (2027.17)^2 \cdot (0.0000000042)^2} = 0.011 \text{ mm}^3$$


El volumen de una arandela individual, expresado con su incertidumbre es:

$$V_1 = \overline{V_1} \pm u_{V_1} = 1183.364 \pm 0.011 \text{ mm}^3$$

```python
u_VII = ((104.16*0.00010)**2 + (46.58*0.00010)**2 + (2027.17*0.0000000042)**2) **0.5 # mm^3
print(f"Incertidumbre volumen método II: {u_VII}")
```

#### 6.2.C CÁLCULO DE LA MASA DE UNA ARANDELA

```python
numero_arandelas = np.array([0, 5, 10, 15, 20, 25, 30])
masa_arandelas = np.array([-1.22, 36.24, 74.42, 116.30, 154.42, 193.26, 231.82])
```

```python
res = stats.linregress(numero_arandelas, masa_arandelas)
print(f"Ordenada en el origen: {res.intercept:.6f} g")
print(f"Error de la ordenada en el origen: {res.intercept_stderr:.6f} g")
print(f"Pendiente (m): {res.slope:.6f} g")
print(f"Error de la pendiente: {res.stderr:.6f}")
```

```python
def lineal(x, pendiente, ordenada_origen): 
    """Dado x calcula y(x) asociado a una correlación linear"""
    return np.multiply(pendiente, x) + ordenada_origen 

plt.plot(numero_arandelas, masa_arandelas, 'x', label='Medidas', color="black")
masas_arandelas_teoricas = lineal(numero_arandelas, res.slope, res.intercept)
plt.plot(numero_arandelas, masas_arandelas_teoricas, '-', label='Regresión')
plt.xlabel("Número de arandelas")
plt.ylabel("Masa (g)")
plt.legend()
plt.show()
```

El error de cero que estamos forzando (aunque apenas se percibe en el gráfico por ser prácticamente despreciable frente al rango de las masas) no influye en la pendiente de la recta de regresión de los $n$ puntos. El valor de la masa promedio por arandela será la pendiente de esta recta de regresión. Por tanto:

$$\overline{m_1} = 7.808g$$

Cuya incertidumbre asociada es $u_{m_1} = 0.040g$ y por tanto:

$$m_1 = \overline{m_1} \pm u_{m_1} = 7.808 \pm 0.040 \text{ g}$$

```python
print(f"Pendiente (m): {res.slope:.3f} \u00B1 {res.stderr:.3f} g")
print(f"Ordenada en el origen (b): {res.intercept:.1f} \u00B1 {res.intercept_stderr:.1f} g")
```

#### 6.2.D CÁLCULO DE LA DENSIDAD Y SUS RESPECTIVAS INCERTIDUMBRES

```python
rho_I = masa / V_I   # g/mm^3
rho_II = masa / V_II # g/mm^3
```

```python
def propagacion_incertidumbre_rho(rho, m, u_m, v, u_v):
    return rho * np.sqrt((u_m / m)**2 + (u_v / v)**2)
```

```python
u_rho_I = propagacion_incertidumbre_rho(rho_I, masa, u_masa, V_I, u_VI)
print(f"ρI = {rho_I} ± {u_rho_I} g/mm^3")
rho_II = masa / V_II
u_rho_II = propagacion_incertidumbre_rho(rho_II, masa, u_masa, V_II, u_VII)
print(f"ρII = {rho_II} ± {u_rho_II} g/mm^3")
```
