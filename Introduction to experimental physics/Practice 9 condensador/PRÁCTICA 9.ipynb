## 1. INTRODUCCIÓN Y OBJETIVOS

<p style="text-indent: 2em; text-align: justify;"> Un circuito eléctrico alimentado por una fuente de corriente continua presenta propiedades muy distintas en presencia de un condensador a las propiedades que mostró en la práctica anterior. Un condensador es un aparato que almacena carga si es conectado a una fuente de voltaje. Conforme el tiempo avanza, almacena cada vez más cantidad de carga, hasta su límite. A partir de ese momento, por la rama del circuito en la que se encuentra el condensador no pasa corriente alguna pues actúa como barrera. 

<p style="text-indent: 2em; text-align: justify;"> Si el circuito se desconecta de la fuente de alimentación y se cierra, el condensador empieza a descargarse. En esta ocasión, la carga acumulada en el condensador disminuye asintóticamente hasta volverse nula y la intensidad vuelve a recuperar su valor en la rama en la que se encontraba el aparato. En experimentos de esta índole, es común incorporar una resistencia al circuito para que el proceso de carga-descarga se demore y se pueda apreciar experimentalmente los cambios en la carga del condensador y el valor de la intensidad. ## 2. MARCO TEÓRICO GENERAL

<p style="text-indent: 2em; text-align: justify;"> La capacidad de un condensador se define como la carga eléctrica que es capaz de almacenar para un determinado voltaje aplicado entre sus extremos:

$$ C = \frac{Q}{V} \tag{1}$$

<p style="text-indent: 2em; text-align: justify;">Si conectamos el condensador en serie a uan resistencia y se alimenta el circuito con una fuente de tensión, se cumple:

$$ V_0 = V_R (t) + V_C (t) \tag{2}$$

<p style="text-indent: 2em; text-align: justify;"> Siendo $V_0$ el potencial de la fuente, $V_R (t)$ el potencial al que está sometida la resistencia y $V_C (t)$ el potencial entre los terminales del condensador. Por la ley de Ohm y la ecuación ($1$):

$$ \frac{V_0 C}{\tau} = \frac{dQ(t)}{dt} + \frac{Q(t)}{\tau} \tag{3}$$

<p style="text-indent: 2em; text-align: justify;"> Donde se ha introducido la variable $\tau = RC$, que llamaremos tiempo característico. Resolviendo la ecuación diferencial y derivando la solución, hallamos $Q(t)$ e $I(t)$, respectivamente:

$$ Q(t) = V_0 C (1 - e^{-t/\tau}) \tag{4}$$

$$ I(t) = \left(\frac{V_0}{R}\right)e^{-t/\tau} \tag{5}$$

<p style="text-indent: 2em; text-align: justify;">Para el voltaje del condensador volvemos a recurrir a la ecuación ($1$):

$$ V_C (t) = V_0 (1-e^{-t/\tau}) \tag{6}$$

<p style="text-indent: 2em; text-align: justify;">Si ahora suponemos que estamos descargando el condensador, obtenemos la siguiente relación matemática:

$$ 0 = V_R (t) + V_C (t) \implies 0 = \frac{dQ(t)}{dt} + \frac{Q(t)}{\tau} \tag{7}$$

<p style="text-indent: 2em; text-align: justify;">Volviendo a resolver ($7$), concluimos que, para el proceso de descarga:

$$Q(t) = V_0 C e^{-t/\tau} \tag{8} $$

$$ V_C (t) = V_0 e^{-t/\tau} \tag{9}$$

#### 4.4 Cálculo de la capacidad del condensador con su incertidumbre

<p style="text-indent: 2em; text-align: justify;">Una vez obtenidos los valores de $C$ del circuito a través de los cuatro ajustes no lineales realizados (dos para el proceso de carga y dos para el de descarga), procedemos a calcular el valor más probable de la capacidad mediante una media ponderada. Este método permite asignar un mayor peso a aquellas medidas que presentan una menor incertidumbre experimental, minimizando así el error del resultado final.

<p style="text-indent: 2em; text-align: justify;">Los valores obtenidos en los apartados anteriores para las capacidades $C_i$ y sus respectivas incertidumbres $u_{C_i}$ se resumen en la siguiente tabla:

| Ajuste | Parámetro | Capacidad ($C_i \pm u_{C_i}$) [$\mu$F] |
| :--- | :--- | :--- |
| Carga - Intensidad | $C_1$ | $157.68 \pm 0.42$ |
| Carga - Voltaje | $C_2$ | $139.36 \pm 0.44$ |
| Descarga - Intensidad | $C_3$ | $118.18 \pm 0.22$ |
| Descarga - Voltaje | $C_4$ | $124.3 \pm 1.4$ |

<p style="text-indent: 2em; text-align: justify;">Para el cálculo de la media ponderada $\langle C \rangle$, utilizamos la expresión:

$$\langle C \rangle = \frac{\sum_{i=1}^{4} w_i C_i}{\sum_{i=1}^{4} w_i} = 130.79 \pm 0.17 \quad \mu F$$

<p style="text-indent: 2em; text-align: justify;">donde los pesos $w_i$ se definen como el inverso del cuadrado de la incertidumbre de cada medida: $w_i = \frac{1}{u_{C_i}^2}$.


##### Deducción de la incertidumbre de la media ponderada

<p style="text-indent: 2em; text-align: justify;">Partiendo de la ley general de propagación de incertidumbres para variables independientes. La expresión general para una función $f$ que depende de diversas variables $x_i$ es:

$$u_f^2 = \sum_{i} \left( \frac{\partial f}{\partial x_i} \right)^2 u_{x_i}^2$$

<p style="text-indent: 2em; text-align: justify;">Aplicando esta ley a nuestra expresión de la media ponderada $\langle C \rangle = f(C_1, C_2, C_3, C_4)$:

$$u_{\langle C \rangle}^2 = \sum_{j=1}^{4} \left( \frac{\partial \langle C \rangle}{\partial C_j} \right)^2 u_{C_j}^2$$

<p style="text-indent: 2em; text-align: justify;">Calculamos la derivada parcial de la media ponderada respecto a una capacidad individual $C_j$:

$$\frac{\partial \langle C \rangle}{\partial C_j} = \frac{\partial}{\partial C_j} \left( \frac{\sum w_i C_i}{\sum w_i} \right) = \frac{w_j}{\sum w_i}$$

<p style="text-indent: 2em; text-align: justify;">Sustituyendo esta derivada en la sumatoria de la ley de propagación:

$$u_{\langle C \rangle}^2 = \sum_{j=1}^{4} \left( \frac{w_j}{\sum w_i} \right)^2 u_{C_j}^2 = \frac{1}{(\sum w_i)^2} \sum_{j=1}^{4} w_j^2 u_{C_j}^2$$

<p style="text-indent: 2em; text-align: justify;">Utilizando la definición del peso $w_j = \frac{1}{u_{C_j}^2}$, podemos reescribir el término $u_{C_j}^2$ como $\frac{1}{w_j}$. Al sustituirlo en la ecuación:

$$u_{\langle C \rangle}^2 = \frac{1}{(\sum w_i)^2} \sum_{j=1}^{4} \left( w_j^2 \cdot \frac{1}{w_j} \right) = \frac{1}{(\sum w_i)^2} \sum_{j=1}^{4} w_j$$

<p style="text-indent: 2em; text-align: justify;">Simplificando los sumatorios obtenemos la expresión final:

$$u_{\langle C \rangle}^2 = \frac{\sum w_i}{(\sum w_i)^2} = \frac{1}{\sum w_i} \implies u_{\langle C \rangle} = \sqrt{\frac{1}{\sum_{i=1}^{4} \frac{1}{u_{C_i}^2}}}$$

<p style="text-indent: 2em; text-align: justify;">Esta última expresión coincide con la ecuación (10) propuesta en el guion de la práctica.

### 7.2 Proceso de carga del condensador

```python
Vc_raw = np.array([-0.39, -0.86, -1.97, -2.86, -3.68, -4.36, -4.91, -5.42, -5.84, -6.21, 
                   -6.51, -7.03, -7.23, -7.39, -7.54, -7.56, -7.76, -7.85, -7.92, -7.98, 
                   -8.03, -8.08]) 

Ic_raw = np.array([-10.3, -9.0, -7.9, -6.9, -6.1, -5.4, -4.8, -4.2, -3.8, -3.4,
                   -3.1, -2.8, -2.6, -2.4, -2.2, -2.1, -1.96, -1.86, -1.77, -1.69, 
                   -1.63, -1.58, -1.53]) 

n_puntos = min(len(Vc_raw), len(Ic_raw))
t_arr = np.array([15*i for i in range(n_puntos)])
Vc_arr = np.abs(Vc_raw[:n_puntos])
Ic_arr = np.abs(Ic_raw[:n_puntos])
Ic_arr = Ic_arr * 1e-6

def modelo_corriente(t, tau, I0):
    return I0 * np.exp(-t / tau)

def modelo_voltaje(t, tau):
    return V0 * (1 - np.exp(-t / tau))

V0 = 9.44
R = 0.8734 * 1e6
```

```python
# AJUSTE CORRIENTE CARGA(Ic)
popt_I, pcov_I = curve_fit(modelo_corriente, t_arr, Ic_arr, p0=[100, 10e-6])
tau1 = popt_I[0]
I0 = popt_I[1]
u_tau1 = np.sqrt(pcov_I[0,0])

# Ajuste Voltaje (Vc)
popt_V, pcov_V = curve_fit(modelo_voltaje, t_arr, Vc_arr, p0=[100])
tau2 = popt_V[0]
u_tau2 = np.sqrt(pcov_V[0,0])

R_ohm = R 

C1 = tau1 / R_ohm
C2 = tau2 / R_ohm
u_C1 = C1 * np.sqrt((u_tau1/tau1)**2 + (u_R/R_ohm)**2)
u_C2 = C2 * np.sqrt((u_tau2/tau2)**2 + (u_R/R_ohm)**2)


print(f"RESULTADOS CARGA:")
print(f"tau (de I): {tau1:.2f} ± {u_tau1:.2f} s  --> C1 = {C1*1e6} ± {u_C1} uF")
print(f"tau (de V): {tau2:.2f} ± {u_tau2:.2f} s  --> C2 = {C2*1e6} ± {u_C2} uF")
```

### 7.4 Cálculo de la capacidad

```python
capacidades = np.array([C1*1e6, C2*1e6, C3, C4])  #  microFaradios
incertidumbres = np.array([u_C1, u_C2, u_C3, u_C4]) 

pesos = 1 / (incertidumbres**2) 
C_media = np.sum(pesos * capacidades) / np.sum(pesos) 
u_C_media = np.sqrt(1 / np.sum(1 / incertidumbres**2))  

for i in range(4):
    c_val, c_err = (capacidades[i], incertidumbres[i])
    print(f"C_{i+1}: ({c_val} ± {c_err}) µF")

c_final, u_final = (C_media, u_C_media)
print(f"<C> = ({c_final} ± {u_final}) µF")
```
