Una vez instalado, ejecuta **Anaconda Navigator** en tu ordenador.

### Instalación de software necesario para exportar ficheros Jupyter Notebook a formato PDF

En el menú principal de Anaconda Navigator, lanza **Jupyter Notebook**.

En la esquina superior derecha, despliega el botón _New_ y abre una _Terminal_.

En la terminal, instala _Pandas_ introduciendo el comando:

```python
pip install pandoc
```

En la terminal, instala _nbconvert_ introduciendo el comando:

```python
pip install nbconvert[webpdf]
```

### Comprobación de la instalación

Abre de nuevo el menú principal de Jupyter Notebook. En la esquina superior derecha, presiona el botón _Upload_ y carga el código de este mismo notebook, "Instalacion.ipynb".

Intenta guardar el Notebook en formato PDF, presionando el botón _File_ --> _Save and Export Notebook As..._ --> _Webpdf_.

Si los programas se han instalado correctamente y has seguido las instrucciones, se debería descargar en la carpeta _Descargas_ de tu ordenador el fichero "Instalacion.pdf", mostrando en formato PDF estas mismas instrucciones.
